### 1.1 Preprocessing Steps

The dataset was loaded from four files: final_posts.json, emotion.csv, data_identification.csv, and sampleSubmission.csv. They were merged using post_id, resulting in 3,360 training samples and 1,440 test samples across six emotion classes.

The data was checked for missing values and duplicate IDs. For traditional machine learning models, text was cleaned by converting it to lowercase, removing URLs and user mentions, and removing extra spaces. For Transformer models, the original text was kept.

An 80/20 stratified split was used for training and validation. An additional holdout set was also created to check model performance. Duplicate texts and conflicting labels were examined, but the original training labels were not changed.

### 1.2 Feature Engineering Steps

For traditional machine learning models, TF-IDF was used to convert text into numerical features. Both unigrams and bigrams were tested.

For Transformer models, pretrained tokenizers were used to convert the original text into input IDs and attention masks, with a maximum length of 128 tokens.

Two pretrained emotion models, GoEmotions and Twitter-RoBERTa-Large, were also used. Their outputs were mapped to the six emotion classes, and the scores were normalized.

Finally, the prediction probabilities from DistilBERT, GoEmotions, and Twitter-RoBERTa-Large were combined using a weighted average to improve classification performance.

### 1.3 Explanation of Your Model

**1. Model Development**

Several models were tested to improve emotion classification performance. Traditional machine learning models using TF-IDF were first used as baselines. After that, DistilBERT was fine-tuned on the labeled training data.

To improve stability, three DistilBERT models were trained with different random seeds (42, 123, and 2026). Their prediction probabilities were averaged to create an ensemble, which achieved a Kaggle Public F1 score of **0.5914**.

**2. Final Ensemble Model**

Two additional pretrained emotion models were tested:

- **GoEmotions:** A RoBERTa-based model that predicts 28 emotion categories.
- **Twitter-RoBERTa-Large:** A model trained on Twitter data that predicts 11 emotion categories.

For both models, the six emotions required by the competition were selected, and their prediction scores were normalized.

The final model combines three components using weighted averaging:

- **DistilBERT Ensemble: 32%**
- **GoEmotions: 48%**
- **Twitter-RoBERTa-Large: 20%**

The emotion with the highest combined score is selected as the final prediction. This approach combines models trained on different data sources to improve classification performance.

**3. Experiments and Results**

Different ensemble combinations and training methods were tested. The main Kaggle results are shown below:

| Model | Kaggle Public F1 |
|---|---:|
| DistilBERT Ensemble | 0.5914 |

Adding GoEmotions gave a clear improvement over DistilBERT alone. Adding Twitter-RoBERTa-Large provided a smaller but consistent improvement in the validation experiments and increased the Kaggle Public F1 to **0.6289**.

Additional experiments included class-specific score adjustments, supervised emotion features, and LoRA fine-tuning. However, these methods were not selected because their improvements were small or inconsistent across validation sets.

For the clean holdout evaluation, a separate DistilBERT ensemble was trained on 2,150 samples, while GoEmotions and Twitter-RoBERTa-Large remained unchanged. After removing overlapping texts, 520 samples were used for evaluation.

## 1. Preprocessing Steps

```python
### Add the code related to the preprocessing steps in cells inside this section


import pandas as pd
import json
import re
from pathlib import Path

# 1. Load competition data
data_dir = Path("data/competition")

with open(data_dir / "final_posts.json", "r", encoding="utf-8") as f:
    posts = json.load(f)

posts_df = pd.DataFrame(posts)
emotion_df = pd.read_csv(data_dir / "emotion.csv")
split_df = pd.read_csv(data_dir / "data_identification.csv")
sample_df = pd.read_csv(data_dir / "sampleSubmission.csv")

# 2. Check dataset information
print("Posts:", posts_df.shape)
print("Emotion:", emotion_df.shape)
print("Split:", split_df.shape)
print("Submission:", sample_df.shape)

print("\nEmotion distribution:")
print(emotion_df["label"].value_counts())

print("\nTrain/Test distribution:")
print(split_df["set"].value_counts())

# 3. Check missing and duplicate IDs
for name, df in [
    ("Posts", posts_df),
    ("Emotion", emotion_df),
    ("Split", split_df)
]:
    print(f"\n{name}:")
    print("Missing IDs:", df["post_id"].isna().sum())
    print("Duplicate IDs:", df["post_id"].duplicated().sum())

# 4. Merge posts, splits, and labels
df = posts_df.merge(
    split_df, on="post_id", how="inner", validate="one_to_one"
)
df = df.merge(
    emotion_df, on="post_id", how="left", validate="one_to_one"
)

train_df = df[df["set"] == "train"].copy()
test_df = df[df["set"] == "test"].copy()

# 5. Check merged data
print("\nMerged shape:", df.shape)
print("Train shape:", train_df.shape)
print("Test shape:", test_df.shape)

print("Missing train labels:", train_df["label"].isna().sum())
print("Missing test labels:", test_df["label"].isna().sum())

test_ids = set(test_df["post_id"])
submission_ids = set(sample_df["post_id"])
print("Submission IDs match test IDs:", test_ids == submission_ids)

# 6. Basic text cleaning
def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"http\S+|www\.\S+", " ", text)
    text = re.sub(r"@\w+", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

train_df["clean_text"] = train_df["text"].fillna("").map(clean_text)
test_df["clean_text"] = test_df["text"].fillna("").map(clean_text)

# 7. Check cleaned text
print("\nSample cleaned tweets:")
print(train_df[["text", "clean_text", "label"]].head(5).to_string(index=False))

print("\nEmpty cleaned texts:")
print("Train:", (train_df["clean_text"] == "").sum())
print("Test:", (test_df["clean_text"] == "").sum())

print("\nPreprocessing completed!")
```

## 2. Feature Engineering Steps

```python
### Add the code related to the feature engineering steps in cells inside this section


from sklearn.feature_extraction.text import TfidfVectorizer

# 1. Prepare text and labels from preprocessing
X_train_text = train_df["clean_text"]
X_test_text = test_df["clean_text"]
y_train = train_df["label"]

# 2. TF-IDF with unigrams (baseline)
tfidf_uni = TfidfVectorizer(
    ngram_range=(1, 1),
    min_df=2,
    sublinear_tf=True
)

X_train_uni = tfidf_uni.fit_transform(X_train_text)
X_test_uni = tfidf_uni.transform(X_test_text)

# 3. TF-IDF with unigrams and bigrams
tfidf_bi = TfidfVectorizer(
    ngram_range=(1, 2),
    min_df=2,
    sublinear_tf=True
)

X_train_bi = tfidf_bi.fit_transform(X_train_text)
X_test_bi = tfidf_bi.transform(X_test_text)

# 4. Display feature dimensions
print("Unigram TF-IDF")
print("Train shape:", X_train_uni.shape)
print("Test shape:", X_test_uni.shape)
print("Vocabulary size:", len(tfidf_uni.vocabulary_))

print("\nUnigram + Bigram TF-IDF")
print("Train shape:", X_train_bi.shape)
print("Test shape:", X_test_bi.shape)
print("Vocabulary size:", len(tfidf_bi.vocabulary_))

print("\nFeature engineering completed!")
```

## 3. Model Implementation Steps

```python

# ==========================================================
# Final V5: Three-Seed DistilBERT Training
#
# Validation mode:
#   Train 2688 samples, validate on 672 samples
#
# Full mode:
#   Train all 3360 samples for Kaggle inference
#
# Training is DISABLED by default.
# Existing trained models will NOT be overwritten.
# ==========================================================

import gc
import numpy as np
import pandas as pd
import torch

from pathlib import Path
from torch.utils.data import Dataset
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    DataCollatorWithPadding,
    TrainingArguments,
    Trainer,
    set_seed
)

# ----------------------------------------------------------
# 1. Configuration
# ----------------------------------------------------------

RUN_V5_TRAINING = False  # Keep False: models already trained
TRAIN_MODE = "full"      # "full" or "validation"

MODEL_NAME = "distilbert/distilbert-base-uncased"
SEEDS = [42, 123, 2026]

EMOTIONS = [
    "anger", "disgust", "fear",
    "joy", "sadness", "surprise"
]

LABEL2ID = {name: i for i, name in enumerate(EMOTIONS)}
ID2LABEL = {i: name for name, i in LABEL2ID.items()}

MAX_LENGTH = 128
EPOCHS = 3
BATCH_SIZE = 8
LEARNING_RATE = 2e-5
WEIGHT_DECAY = 0.01

# Separate reproduction directory: never overwrite V5
OUTPUT_ROOT = (
    Path("v5_reproduction_only") / TRAIN_MODE
)

assert TRAIN_MODE in ["full", "validation"]
assert len(train_df) == 3360

# ----------------------------------------------------------
# 2. Prepare datasets
# ----------------------------------------------------------

def encode_v5_labels(series):
    if pd.api.types.is_numeric_dtype(series):
        labels = series.to_numpy(dtype=int)
    else:
        labels = (
            series.astype(str)
            .str.lower()
            .str.strip()
            .map(LABEL2ID)
            .to_numpy()
        )

    assert np.isin(labels, range(6)).all()
    return labels.astype(np.int64)


class V5Dataset(Dataset):

    def __init__(self, dataframe, tokenizer):

        self.encodings = tokenizer(
            dataframe["text"]
            .fillna("")
            .astype(str)
            .tolist(),
            truncation=True,
            max_length=MAX_LENGTH
        )

        self.labels = encode_v5_labels(
            dataframe["label"]
        )

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        item = {
            key: values[idx]
            for key, values in self.encodings.items()
        }
        item["labels"] = int(self.labels[idx])
        return item


def prepare_v5_splits(mode):

    if mode == "validation":
        dev_df, val_df = train_test_split(
            train_df,
            test_size=0.2,
            random_state=42,
            stratify=train_df["label"]
        )
        return (
            dev_df.reset_index(drop=True),
            val_df.reset_index(drop=True)
        )

    return train_df.reset_index(drop=True), None


# ----------------------------------------------------------
# 3. Training function
# ----------------------------------------------------------

def train_v5_three_seeds(mode):

    assert torch.cuda.is_available(), "CUDA unavailable."

    # Protect existing experiment results
    assert not OUTPUT_ROOT.exists(), (
        f"{OUTPUT_ROOT} already exists. "
        "Existing files will not be overwritten."
    )

    train_data, val_data = prepare_v5_splits(mode)

    print("Training samples:", len(train_data))
    print(
        "Validation samples:",
        0 if val_data is None else len(val_data)
    )

    tokenizer = AutoTokenizer.from_pretrained(
        MODEL_NAME
    )

    train_dataset = V5Dataset(train_data, tokenizer)

    val_dataset = (
        V5Dataset(val_data, tokenizer)
        if val_data is not None
        else None
    )

    collator = DataCollatorWithPadding(
        tokenizer=tokenizer
    )

    all_validation_probs = []

    OUTPUT_ROOT.mkdir(
        parents=True,
        exist_ok=False
    )

    for seed in SEEDS:

        print(f"\n========== V5 SEED {seed} ==========")

        set_seed(seed)

        model = (
            AutoModelForSequenceClassification
            .from_pretrained(
                MODEL_NAME,
                num_labels=6,
                id2label=ID2LABEL,
                label2id=LABEL2ID,
                dtype=torch.float32
            )
        )

        args = TrainingArguments(
            output_dir=str(
                OUTPUT_ROOT / f"trainer_seed_{seed}"
            ),
            learning_rate=LEARNING_RATE,
            num_train_epochs=EPOCHS,
            per_device_train_batch_size=BATCH_SIZE,
            per_device_eval_batch_size=BATCH_SIZE,
            gradient_accumulation_steps=1,
            weight_decay=WEIGHT_DECAY,
            lr_scheduler_type="linear",
            warmup_steps=0,
            fp16=True,
            eval_strategy="no",
            save_strategy="no",
            logging_strategy="epoch",
            seed=seed,
            data_seed=seed,
            report_to="none"
        )

        trainer = Trainer(
            model=model,
            args=args,
            train_dataset=train_dataset,
            processing_class=tokenizer,
            data_collator=collator
        )

        trainer.train()

        # Save trained model and tokenizer
        model_dir = OUTPUT_ROOT / f"seed_{seed}"
        trainer.save_model(str(model_dir))
        tokenizer.save_pretrained(str(model_dir))

        # Validation mode: save prediction probabilities
        if val_dataset is not None:

            output = trainer.predict(val_dataset)

            logits = output.predictions
            if isinstance(logits, tuple):
                logits = logits[0]

            logits = logits.astype(np.float64)
            logits -= logits.max(
                axis=1, keepdims=True
            )

            probs = np.exp(logits)
            probs /= probs.sum(
                axis=1, keepdims=True
            )

            all_validation_probs.append(probs)

            np.save(
                OUTPUT_ROOT /
                f"validation_probs_seed_{seed}.npy",
                probs
            )

        del trainer, model
        gc.collect()
        torch.cuda.empty_cache()

    # Three-seed validation soft voting
    if val_dataset is not None:

        ensemble_probs = np.mean(
            all_validation_probs,
            axis=0
        )

        y_true = encode_v5_labels(
            val_data["label"]
        )

        f1 = f1_score(
            y_true,
            ensemble_probs.argmax(axis=1),
            labels=range(6),
            average="macro",
            zero_division=0
        )

        print("\nValidation Macro F1:", round(f1, 6))

        np.save(
            OUTPUT_ROOT / "ensemble_probs.npy",
            ensemble_probs
        )

    print("\nV5 training completed.")


# ----------------------------------------------------------
# 4. Safety gate
# ----------------------------------------------------------

print("========== V5 TRAINING CONFIG ==========")
print("Base model:", MODEL_NAME)
print("Seeds:", SEEDS)
print("Mode:", TRAIN_MODE)
print("Epochs:", EPOCHS)
print("Batch size:", BATCH_SIZE)
print("Learning rate:", LEARNING_RATE)
print("Max length:", MAX_LENGTH)

if RUN_V5_TRAINING:
    train_v5_three_seeds(TRAIN_MODE)
else:
    print("\nTraining skipped: existing V5 models preserved.")
    print("Expected existing models:")
    for seed in SEEDS:
        model_dir = Path(
            f"models/distilbert_v5/seed_{seed}"
        )
        print(
            f"Seed {seed}:",
            "FOUND" if (
                model_dir / "model.safetensors"
            ).exists() else "NOT FOUND"
        )
```

```python

# ==========================================================
# Final Model: Reproduce Kaggle Submission
#
# DistilBERT 3 Seeds: 32%
# GoEmotions:         48%
# Twitter-RoBERTa:    20%
#
# Inference only. No training.
# Never overwrite the original submission.
# ==========================================================

import gc
import numpy as np
import pandas as pd
import torch

from pathlib import Path
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification
)

FINAL_EMOTIONS = np.array([
    "anger", "disgust", "fear",
    "joy", "sadness", "surprise"
])

FINAL_SEEDS = [42, 123, 2026]
FINAL_MAX_LENGTH = 128

REFERENCE_CSV = Path(
    "submission_v5_go_twitter_large_32_48_20.csv"
)

REPRODUCED_CSV = Path(
    "submission_final_reproduced_check.csv"
)

assert torch.cuda.is_available()
assert REFERENCE_CSV.exists()

device = torch.device("cuda")

print("GPU:", torch.cuda.get_device_name(0))

# ----------------------------------------------------------
# 1. Align test data with sample submission
# ----------------------------------------------------------

id_col = (
    "post_id"
    if "post_id" in sample_df.columns
    else "id"
)

final_ids = sample_df[id_col].astype(str).to_numpy()

test_lookup = test_df.copy()
test_lookup["post_id"] = (
    test_lookup["post_id"].astype(str)
)

assert len(final_ids) == 1440
assert len(set(final_ids)) == 1440
assert test_lookup["post_id"].is_unique
assert set(final_ids) == set(test_lookup["post_id"])

final_texts = (
    test_lookup
    .set_index("post_id")
    .loc[final_ids, "text"]
    .fillna("")
    .astype(str)
    .tolist()
)

print("Test alignment verified:", len(final_texts))

# ----------------------------------------------------------
# 2. Reusable inference function
# ----------------------------------------------------------

def final_predict_logits(
    model,
    tokenizer,
    texts,
    batch_size
):
    model.eval()
    outputs = []

    with torch.inference_mode():

        for start in range(
            0, len(texts), batch_size
        ):

            batch = texts[
                start:start + batch_size
            ]

            inputs = tokenizer(
                batch,
                padding=True,
                truncation=True,
                max_length=FINAL_MAX_LENGTH,
                return_tensors="pt"
            ).to(device)

            logits = model(**inputs).logits

            outputs.append(
                logits.float().cpu().numpy()
            )

    return np.concatenate(outputs, axis=0)


def normalize_six_sigmoid(
    logits,
    model_config
):
    names = {
        str(name).lower().strip(): int(idx)
        for idx, name in model_config.id2label.items()
    }

    assert all(
        emotion in names
        for emotion in FINAL_EMOTIONS
    )

    indices = [
        names[emotion]
        for emotion in FINAL_EMOTIONS
    ]

    scores = torch.sigmoid(
        torch.from_numpy(logits)
    ).numpy()[:, indices]

    return scores / np.maximum(
        scores.sum(axis=1, keepdims=True),
        1e-12
    )


# ----------------------------------------------------------
# 3. DistilBERT three-seed ensemble
# ----------------------------------------------------------

print("\n========== DISTILBERT ==========")

v5_seed_probs = []

for seed in FINAL_SEEDS:

    model_dir = Path(
        f"models/distilbert_v5/seed_{seed}"
    )

    assert (
        model_dir / "model.safetensors"
    ).exists(), model_dir

    tokenizer = AutoTokenizer.from_pretrained(
        str(model_dir),
        local_files_only=True
    )

    model = (
        AutoModelForSequenceClassification
        .from_pretrained(
            str(model_dir),
            local_files_only=True
        )
        .to(device)
    )

    assert model.config.label2id == {
        emotion: i
        for i, emotion in enumerate(FINAL_EMOTIONS)
    }

    logits = final_predict_logits(
        model,
        tokenizer,
        final_texts,
        batch_size=16
    )

    probs = torch.softmax(
        torch.from_numpy(logits),
        dim=-1
    ).numpy()

    v5_seed_probs.append(probs)

    print(f"Seed {seed}: completed")

    del model, tokenizer, logits, probs
    gc.collect()
    torch.cuda.empty_cache()

v5_probs = np.mean(
    np.stack(v5_seed_probs),
    axis=0
)

# ----------------------------------------------------------
# 4. GoEmotions pretrained model
# ----------------------------------------------------------

print("\n========== GOEMOTIONS ==========")

go_name = "SamLowe/roberta-base-go_emotions"

go_tokenizer = AutoTokenizer.from_pretrained(
    go_name
)

go_model = (
    AutoModelForSequenceClassification
    .from_pretrained(go_name)
    .to(device)
)

go_logits = final_predict_logits(
    go_model,
    go_tokenizer,
    final_texts,
    batch_size=16
)

go_probs = normalize_six_sigmoid(
    go_logits,
    go_model.config
)

print("GoEmotions: completed")

del go_model, go_tokenizer, go_logits
gc.collect()
torch.cuda.empty_cache()

# ----------------------------------------------------------
# 5. Twitter-RoBERTa-Large pretrained model
# ----------------------------------------------------------

print("\n========== TWITTER-ROBERTA ==========")

twitter_name = (
    "cardiffnlp/"
    "twitter-roberta-large-emotion-latest"
)

twitter_tokenizer = (
    AutoTokenizer.from_pretrained(twitter_name)
)

twitter_model = (
    AutoModelForSequenceClassification
    .from_pretrained(twitter_name)
    .to(device=device, dtype=torch.float16)
)

twitter_logits = final_predict_logits(
    twitter_model,
    twitter_tokenizer,
    final_texts,
    batch_size=4
)

twitter_probs = normalize_six_sigmoid(
    twitter_logits,
    twitter_model.config
)

print("Twitter-RoBERTa-Large: completed")

del twitter_model, twitter_tokenizer, twitter_logits
gc.collect()
torch.cuda.empty_cache()

# ----------------------------------------------------------
# 6. Final weighted ensemble
# ----------------------------------------------------------

assert (
    v5_probs.shape
    == go_probs.shape
    == twitter_probs.shape
    == (1440, 6)
)

assert all(
    np.isfinite(p).all()
    for p in [v5_probs, go_probs, twitter_probs]
)

final_probs = (
    0.32 * v5_probs
    + 0.48 * go_probs
    + 0.20 * twitter_probs
)

final_labels = FINAL_EMOTIONS[
    final_probs.argmax(axis=1)
]

submission = pd.DataFrame({
    "id": final_ids,
    "label": final_labels
})

assert submission.shape == (1440, 2)
assert list(submission.columns) == ["id", "label"]
assert not submission.isna().any().any()

# ----------------------------------------------------------
# 7. Compare against actual submitted CSV
# ----------------------------------------------------------

reference = pd.read_csv(
    REFERENCE_CSV,
    dtype={"id": str}
)

assert list(reference.columns) == ["id", "label"]
assert np.array_equal(
    reference["id"].to_numpy(),
    submission["id"].to_numpy()
)

mismatch = (
    reference["label"].to_numpy()
    != submission["label"].to_numpy()
)

mismatch_count = int(mismatch.sum())

print("\n========== REPRODUCTION CHECK ==========")

print("Reference submission:", REFERENCE_CSV)
print("Total test samples:", len(submission))
print("Different predictions:", mismatch_count)
print(
    "Agreement:",
    f"{1 - mismatch_count / 1440:.4%}"
)

if mismatch_count == 0:

    if REPRODUCED_CSV.exists():

        existing = pd.read_csv(
            REPRODUCED_CSV,
            dtype={"id": str}
        )

        assert existing.equals(submission)
        print("Existing reproduced CSV verified.")

    else:

        submission.to_csv(
            REPRODUCED_CSV,
            index=False,
            encoding="utf-8"
        )

        print("Reproduced CSV saved:", REPRODUCED_CSV)

    print("SUCCESS: All predictions match.")

else:

    differences = pd.DataFrame({
        "id": final_ids[mismatch],
        "original": reference.loc[
            mismatch, "label"
        ].to_numpy(),
        "reproduced": final_labels[mismatch]
    })

    print("\nPrediction differences:")
    print(differences.head(10).to_string(index=False))

    print("\nExact reproduction not verified.")
    print("No new submission CSV saved.")

print("\nOriginal Kaggle 0.6289 CSV preserved.")
print("No model training performed.")
```

```python

# ==========================================================
# Experiment 20: DistilBERT Seed 777
# Validation + 4-Seed Ensemble
# No Kaggle submission
# ==========================================================

import json
import numpy as np
import pandas as pd
import torch

from pathlib import Path
from torch.utils.data import Dataset
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    f1_score,
    accuracy_score,
    precision_recall_fscore_support
)
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    DataCollatorWithPadding,
    TrainingArguments,
    Trainer,
    set_seed
)

# ----------------------------------------------------------
# 1. Settings
# ----------------------------------------------------------

MODEL_NAME = "distilbert/distilbert-base-uncased"

SEED = 777
MAX_LENGTH = 128
BATCH_SIZE = 8
EPOCHS = 3
LEARNING_RATE = 2e-5
WEIGHT_DECAY = 0.01

OUT_DIR = Path("distilbert_seed777_validation")
MODEL_DIR = OUT_DIR / "final_model"

if OUT_DIR.exists():
    raise FileExistsError(
        f"{OUT_DIR} already exists. Existing results preserved."
    )

if not torch.cuda.is_available():
    raise RuntimeError("CUDA unavailable.")

print("GPU:", torch.cuda.get_device_name(0))
print("Seed:", SEED)

set_seed(SEED)

# ----------------------------------------------------------
# 2. Original development / validation split
# ----------------------------------------------------------

assert len(train_df) == 3360

dev_777, val_777 = train_test_split(
    train_df,
    test_size=0.2,
    random_state=42,
    stratify=train_df["label"]
)

dev_777 = dev_777.reset_index(drop=True)
val_777 = val_777.reset_index(drop=True)

EMOTIONS_777 = [
    "anger", "disgust", "fear",
    "joy", "sadness", "surprise"
]

label2id = {
    name: i for i, name in enumerate(EMOTIONS_777)
}

id2label = {
    i: name for name, i in label2id.items()
}

def encode_labels(series):
    if pd.api.types.is_numeric_dtype(series):
        y = series.to_numpy(dtype=int)
    else:
        y = (
            series.astype(str)
            .str.strip()
            .str.lower()
            .map(label2id)
            .to_numpy()
        )

    assert np.isin(y, range(6)).all()
    return y.astype(np.int64)

y_train = encode_labels(dev_777["label"])
y_val = encode_labels(val_777["label"])

print("Training:", len(dev_777))
print("Validation:", len(val_777))

# ----------------------------------------------------------
# 3. Load original V5 probabilities
# ----------------------------------------------------------

V5_DIR = Path("distilbert_ensemble_validation")

p42 = np.load(
    V5_DIR / "validation_probs_seed_42.npy"
)

p123 = np.load(
    V5_DIR / "validation_probs_seed_123.npy"
)

p2026 = np.load(
    V5_DIR / "validation_probs_seed_2026.npy"
)

v5_probs = (p42 + p123 + p2026) / 3

assert v5_probs.shape == (672, 6)

v5_pred = v5_probs.argmax(axis=1)

v5_f1 = f1_score(
    y_val,
    v5_pred,
    average="macro",
    labels=range(6),
    zero_division=0
)

assert abs(v5_f1 - 0.620562) < 1e-5, (
    "V5 validation alignment mismatch."
)

print("Original V5 verified:", round(v5_f1, 6))

# ----------------------------------------------------------
# 4. Tokenize original text
# ----------------------------------------------------------

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME
)

class EmotionDataset(Dataset):

    def __init__(self, texts, labels):

        self.encodings = tokenizer(
            texts,
            truncation=True,
            max_length=MAX_LENGTH
        )

        self.labels = labels

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):

        item = {
            k: v[idx]
            for k, v in self.encodings.items()
        }

        item["labels"] = int(self.labels[idx])
        return item

train_dataset = EmotionDataset(
    dev_777["text"].fillna("").astype(str).tolist(),
    y_train
)

val_dataset = EmotionDataset(
    val_777["text"].fillna("").astype(str).tolist(),
    y_val
)

collator = DataCollatorWithPadding(tokenizer)

# ----------------------------------------------------------
# 5. Initialize model
# ----------------------------------------------------------

set_seed(SEED)

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=6,
    id2label=id2label,
    label2id=label2id,
    dtype=torch.float32
)

# ----------------------------------------------------------
# 6. Evaluation metrics
# ----------------------------------------------------------

def compute_metrics(eval_pred):

    logits = eval_pred.predictions

    if isinstance(logits, tuple):
        logits = logits[0]

    predictions = logits.argmax(axis=1)

    return {
        "accuracy": accuracy_score(
            eval_pred.label_ids,
            predictions
        ),
        "macro_f1": f1_score(
            eval_pred.label_ids,
            predictions,
            average="macro",
            labels=range(6),
            zero_division=0
        )
    }

# ----------------------------------------------------------
# 7. Training arguments
# ----------------------------------------------------------

training_args = TrainingArguments(
    output_dir=str(OUT_DIR / "trainer_output"),

    learning_rate=LEARNING_RATE,
    num_train_epochs=EPOCHS,

    per_device_train_batch_size=BATCH_SIZE,
    per_device_eval_batch_size=BATCH_SIZE,

    gradient_accumulation_steps=1,
    weight_decay=WEIGHT_DECAY,

    lr_scheduler_type="linear",
    warmup_steps=0,

    fp16=True,

    eval_strategy="epoch",
    save_strategy="no",
    logging_strategy="epoch",

    load_best_model_at_end=False,

    seed=SEED,
    data_seed=SEED,

    report_to="none"
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=val_dataset,
    processing_class=tokenizer,
    data_collator=collator,
    compute_metrics=compute_metrics
)

# ----------------------------------------------------------
# 8. Train Seed 777
# ----------------------------------------------------------

print("\n========== TRAINING SEED 777 ==========")

trainer.train()

# Use the final epoch, consistent with fixed 3-epoch training.
prediction_output = trainer.predict(val_dataset)

logits = prediction_output.predictions

if isinstance(logits, tuple):
    logits = logits[0]

p777 = torch.softmax(
    torch.tensor(logits, dtype=torch.float32),
    dim=-1
).numpy()

assert p777.shape == (672, 6)

# ----------------------------------------------------------
# 9. Four-seed soft voting
# ----------------------------------------------------------

four_seed_probs = (
    p42 + p123 + p2026 + p777
) / 4.0

seed777_pred = p777.argmax(axis=1)
four_seed_pred = four_seed_probs.argmax(axis=1)

def evaluate(pred):

    return {
        "Macro F1": f1_score(
            y_val,
            pred,
            labels=range(6),
            average="macro",
            zero_division=0
        ),
        "Accuracy": accuracy_score(y_val, pred)
    }

comparison = pd.DataFrame({
    "V5 3-Seed Ensemble": evaluate(v5_pred),
    "Seed 777 Single": evaluate(seed777_pred),
    "4-Seed Ensemble": evaluate(four_seed_pred)
}).T

print("\n========== MODEL COMPARISON ==========")
print(comparison.round(6).to_string())

print(
    "\n4-Seed Delta F1 vs V5:",
    f"{comparison.loc['4-Seed Ensemble', 'Macro F1'] - v5_f1:+.6f}"
)

# ----------------------------------------------------------
# 10. Per-class F1
# ----------------------------------------------------------

def class_f1(pred):
    return precision_recall_fscore_support(
        y_val,
        pred,
        labels=range(6),
        zero_division=0
    )[2]

v5_class = class_f1(v5_pred)
seed777_class = class_f1(seed777_pred)
four_class = class_f1(four_seed_pred)

class_df = pd.DataFrame({
    "Emotion": EMOTIONS_777,
    "V5 F1": v5_class,
    "Seed 777 F1": seed777_class,
    "4-Seed F1": four_class,
    "Delta vs V5": four_class - v5_class
})

print("\n========== PER-CLASS REPORT ==========")
print(class_df.round(4).to_string(index=False))

# ----------------------------------------------------------
# 11. Prediction complementarity
# ----------------------------------------------------------

v5_correct = v5_pred == y_val
four_correct = four_seed_pred == y_val

changed = int(
    np.sum(v5_pred != four_seed_pred)
)

fixed = int(
    np.sum(~v5_correct & four_correct)
)

broken = int(
    np.sum(v5_correct & ~four_correct)
)

different_wrong = int(
    np.sum(
        ~v5_correct &
        ~four_correct &
        (v5_pred != four_seed_pred)
    )
)

print("\n========== V5 COMPLEMENTARITY ==========")

print("Changed predictions:", changed)
print("V5 Wrong -> 4-Seed Correct:", fixed)
print("V5 Correct -> 4-Seed Wrong:", broken)
print("Wrong -> Different Wrong:", different_wrong)
print("Net correct difference:", fixed - broken)

# ----------------------------------------------------------
# 12. Save new model and results
# ----------------------------------------------------------

OUT_DIR.mkdir(parents=True, exist_ok=True)

trainer.save_model(str(MODEL_DIR))
tokenizer.save_pretrained(str(MODEL_DIR))

np.save(
    OUT_DIR / "validation_probs_seed777.npy",
    p777
)

np.save(
    OUT_DIR / "validation_probs_4seed.npy",
    four_seed_probs
)

comparison.to_csv(
    OUT_DIR / "model_comparison.csv"
)

class_df.to_csv(
    OUT_DIR / "per_class_report.csv",
    index=False
)

pd.DataFrame(
    trainer.state.log_history
).to_csv(
    OUT_DIR / "training_history.csv",
    index=False
)

summary = {
    "model": MODEL_NAME,
    "seed": SEED,
    "training_samples": 2688,
    "validation_samples": 672,
    "epochs": EPOCHS,
    "learning_rate": LEARNING_RATE,
    "batch_size": BATCH_SIZE,
    "weight_decay": WEIGHT_DECAY,
    "max_length": MAX_LENGTH,
    "v5_macro_f1": float(v5_f1),
    "seed777_macro_f1": float(
        comparison.loc["Seed 777 Single", "Macro F1"]
    ),
    "four_seed_macro_f1": float(
        comparison.loc["4-Seed Ensemble", "Macro F1"]
    )
}

with open(
    OUT_DIR / "experiment_summary.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        summary,
        f,
        indent=2,
        ensure_ascii=False
    )

print("\n========== EXPERIMENT COMPLETED ==========")
print("Results saved:", OUT_DIR)
print("Original V5 models unchanged.")
print("No Kaggle submission generated.")
```

```python

# ==========================================================
# Experiment 23: GoEmotions Pretrained Model
# Zero-shot 6-emotion diagnostic
# No training / No Kaggle submission
# ==========================================================

import numpy as np
import pandas as pd
import torch

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    f1_score,
    accuracy_score,
    precision_recall_fscore_support
)

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification
)

MODEL_NAME = "SamLowe/roberta-base-go_emotions"

EMOTIONS_GO = [
    "anger", "disgust", "fear",
    "joy", "sadness", "surprise"
]

assert torch.cuda.is_available(), "CUDA unavailable."

device = torch.device("cuda")

# 1. Same 672-sample validation split
_, val_go = train_test_split(
    train_df,
    test_size=0.2,
    random_state=42,
    stratify=train_df["label"]
)

val_go = val_go.reset_index(drop=True)

label2id = {name: i for i, name in enumerate(EMOTIONS_GO)}

if pd.api.types.is_numeric_dtype(val_go["label"]):
    y_true = val_go["label"].to_numpy(dtype=int)
else:
    y_true = (
        val_go["label"]
        .astype(str)
        .str.lower()
        .str.strip()
        .map(label2id)
        .to_numpy(dtype=int)
    )

# 2. Load V5 baseline
v5_dir = "distilbert_ensemble_validation"

v5_probs = np.mean([
    np.load(f"{v5_dir}/validation_probs_seed_{s}.npy")
    for s in [42, 123, 2026]
], axis=0)

v5_pred = v5_probs.argmax(axis=1)

v5_f1 = f1_score(
    y_true,
    v5_pred,
    labels=range(6),
    average="macro"
)

assert abs(v5_f1 - 0.620562) < 1e-5

print("Original V5 verified:", round(v5_f1, 6))

# 3. Load emotion-pretrained RoBERTa
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME
)

model.to(device)
model.eval()

# 4. Identify exact class indices
model_labels = {
    str(name).lower(): int(idx)
    for idx, name in model.config.id2label.items()
}

assert all(name in model_labels for name in EMOTIONS_GO)

selected_indices = [
    model_labels[name] for name in EMOTIONS_GO
]

print("\nSelected emotion indices:", selected_indices)

# 5. Predict original validation texts
texts = val_go["text"].fillna("").astype(str).tolist()

all_scores = []

with torch.inference_mode():

    for start in range(0, len(texts), 16):

        batch = texts[start:start + 16]

        inputs = tokenizer(
            batch,
            padding=True,
            truncation=True,
            max_length=128,
            return_tensors="pt"
        ).to(device)

        logits = model(**inputs).logits

        # Original model uses 28 independent labels.
        # Sigmoid scores are NOT six-class probabilities.
        scores = torch.sigmoid(
            logits.float()
        )[:, selected_indices]

        all_scores.append(scores.cpu().numpy())

scores_go = np.concatenate(all_scores, axis=0)

assert scores_go.shape == (672, 6)

pred_go = scores_go.argmax(axis=1)

# 6. Evaluation
go_f1 = f1_score(
    y_true,
    pred_go,
    labels=range(6),
    average="macro",
    zero_division=0
)

comparison = pd.DataFrame([
    {
        "Model": "V5 DistilBERT Ensemble",
        "Macro F1": v5_f1,
        "Accuracy": accuracy_score(y_true, v5_pred)
    },
    {
        "Model": "GoEmotions Pretrained (No Fine-tuning)",
        "Macro F1": go_f1,
        "Accuracy": accuracy_score(y_true, pred_go)
    }
])

print("\n========== MODEL COMPARISON ==========")
print(comparison.round(6).to_string(index=False))

# 7. Per-class comparison
_, _, v5_class, _ = precision_recall_fscore_support(
    y_true, v5_pred,
    labels=range(6),
    zero_division=0
)

_, _, go_class, _ = precision_recall_fscore_support(
    y_true, pred_go,
    labels=range(6),
    zero_division=0
)

per_class = pd.DataFrame({
    "Emotion": EMOTIONS_GO,
    "V5 F1": v5_class,
    "GoEmotions F1": go_class,
    "Difference": go_class - v5_class
})

print("\n========== PER-CLASS REPORT ==========")
print(per_class.round(4).to_string(index=False))

# 8. Complementarity
v5_correct = v5_pred == y_true
go_correct = pred_go == y_true

print("\n========== V5 COMPLEMENTARITY ==========")

print("Changed:", int((v5_pred != pred_go).sum()))
print(
    "V5 Wrong -> GoEmotions Correct:",
    int((~v5_correct & go_correct).sum())
)
print(
    "V5 Correct -> GoEmotions Wrong:",
    int((v5_correct & ~go_correct).sum())
)

print("\nDiagnostic completed.")
print("No training. No Kaggle submission.")
```

```python

# ==========================================================
# Experiment 25: V5 + GoEmotions Bootstrap Verification
# No training / No inference / No submission
# ==========================================================

import numpy as np
import pandas as pd
from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score, accuracy_score
from scipy.stats import binomtest

# 1. Load saved validation probabilities
v5_dir = Path("distilbert_ensemble_validation")
blend_dir = Path("v5_goemotions_50_50_analysis")

v5_probs = np.mean([
    np.load(v5_dir / f"validation_probs_seed_{s}.npy")
    for s in [42, 123, 2026]
], axis=0)

blend_probs = np.load(
    blend_dir / "combined_validation_probs.npy"
)

assert v5_probs.shape == blend_probs.shape == (672, 6)

# 2. Restore exact validation labels
_, val_check = train_test_split(
    train_df,
    test_size=0.2,
    random_state=42,
    stratify=train_df["label"]
)

emotions = [
    "anger", "disgust", "fear",
    "joy", "sadness", "surprise"
]

label_map = {name: i for i, name in enumerate(emotions)}

if pd.api.types.is_numeric_dtype(val_check["label"]):
    y = val_check["label"].to_numpy(dtype=int)
else:
    y = (
        val_check["label"]
        .astype(str).str.lower().str.strip()
        .map(label_map)
        .to_numpy(dtype=int)
    )

assert np.isin(y, range(6)).all()

v5_pred = v5_probs.argmax(axis=1)
blend_pred = blend_probs.argmax(axis=1)

def macro_f1(true, pred):
    return f1_score(
        true, pred,
        labels=range(6),
        average="macro",
        zero_division=0
    )

v5_f1 = macro_f1(y, v5_pred)
blend_f1 = macro_f1(y, blend_pred)

assert abs(v5_f1 - 0.620562) < 1e-5
assert abs(blend_f1 - 0.658051) < 1e-5

# 3. Paired stratified bootstrap
rng = np.random.default_rng(2026)
N_BOOTSTRAP = 5000

class_indices = [
    np.flatnonzero(y == c)
    for c in range(6)
]

deltas = np.empty(N_BOOTSTRAP)

for b in range(N_BOOTSTRAP):
    indices = np.concatenate([
        rng.choice(idx, size=len(idx), replace=True)
        for idx in class_indices
    ])

    deltas[b] = (
        macro_f1(y[indices], blend_pred[indices])
        - macro_f1(y[indices], v5_pred[indices])
    )

ci_low, ci_high = np.quantile(
    deltas, [0.025, 0.975]
)

# 4. Compare changed predictions
v5_correct = v5_pred == y
blend_correct = blend_pred == y

fixed = int((~v5_correct & blend_correct).sum())
broken = int((v5_correct & ~blend_correct).sum())

# McNemar's exact test for accuracy differences
mcnemar_p = binomtest(
    min(fixed, broken),
    n=fixed + broken,
    p=0.5,
    alternative="two-sided"
).pvalue

# 5. Print
print("========== BOOTSTRAP COMPARISON ==========")
print("V5 Macro F1:", round(v5_f1, 6))
print("Combined Macro F1:", round(blend_f1, 6))
print("Observed Delta:", round(blend_f1 - v5_f1, 6))
print("95% CI:", (round(ci_low, 6), round(ci_high, 6)))
print("Bootstrap Delta > 0:", round(float((deltas > 0).mean()), 4))

print("\n========== PREDICTION CHANGES ==========")
print("Wrong -> Correct:", fixed)
print("Correct -> Wrong:", broken)
print("Net Correct:", fixed - broken)
print("Exact McNemar p-value:", round(mcnemar_p, 6))

print("\nNote: Reusing the validation set for model")
print("selection can make these results optimistic.")
print("No training or Kaggle submission performed.")
```

```python

# ==========================================================
# Experiment 26: V5 + GoEmotions 50/50 Kaggle Candidate
# No training / No automatic submission
# ==========================================================

import gc
import json
import numpy as np
import pandas as pd
import torch

from pathlib import Path
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification
)

# ----------------------------------------------------------
# 1. Settings
# ----------------------------------------------------------

EMOTIONS = [
    "anger", "disgust", "fear",
    "joy", "sadness", "surprise"
]

GO_MODEL = "SamLowe/roberta-base-go_emotions"

SEEDS = [42, 123, 2026]
MAX_LENGTH = 128
BATCH_SIZE = 16

OUT_DIR = Path("v5_goemotions_50_50_test")
SUBMISSION_PATH = Path(
    "submission_v5_goemotions_50_50.csv"
)

if OUT_DIR.exists() or SUBMISSION_PATH.exists():
    raise FileExistsError(
        "Output already exists. Existing files preserved."
    )

assert torch.cuda.is_available(), "CUDA unavailable."
assert len(test_df) == 1440
assert len(sample_df) == 1440

device = torch.device("cuda")

print("GPU:", torch.cuda.get_device_name(0))

# ----------------------------------------------------------
# 2. Align test text to sample submission
# ----------------------------------------------------------

id_col = (
    "post_id" if "post_id" in sample_df.columns
    else "id"
)

sample_ids = sample_df[id_col].astype(str).tolist()

test_lookup = test_df.copy()
test_lookup["post_id"] = (
    test_lookup["post_id"].astype(str)
)

assert test_lookup["post_id"].is_unique
assert len(set(sample_ids)) == 1440
assert set(sample_ids) == set(test_lookup["post_id"])

test_texts = (
    test_lookup.set_index("post_id")
    .loc[sample_ids, "text"]
    .fillna("")
    .astype(str)
    .tolist()
)

print("Test alignment verified:", len(test_texts))

# ----------------------------------------------------------
# 3. Reusable inference function
# ----------------------------------------------------------

def get_logits(model, tokenizer, texts):
    outputs = []

    model.to(device)
    model.eval()

    with torch.inference_mode():
        for start in range(0, len(texts), BATCH_SIZE):

            batch = texts[start:start + BATCH_SIZE]

            inputs = tokenizer(
                batch,
                padding=True,
                truncation=True,
                max_length=MAX_LENGTH,
                return_tensors="pt"
            ).to(device)

            logits = model(**inputs).logits

            outputs.append(
                logits.float().cpu().numpy()
            )

    return np.concatenate(outputs, axis=0)

# ----------------------------------------------------------
# 4. Predict with original V5 three models
# ----------------------------------------------------------

print("\n========== V5 TEST PREDICTION ==========")

v5_seed_probs = []

for seed in SEEDS:

    model_dir = Path(
        f"models/distilbert_v5/seed_{seed}"
    )

    assert model_dir.exists(), model_dir

    tokenizer = AutoTokenizer.from_pretrained(
        str(model_dir),
        local_files_only=True
    )

    model = (
        AutoModelForSequenceClassification
        .from_pretrained(
            str(model_dir),
            local_files_only=True
        )
    )

    assert model.config.label2id == {
        name: i for i, name in enumerate(EMOTIONS)
    }

    logits = get_logits(
        model, tokenizer, test_texts
    )

    probs = torch.softmax(
        torch.from_numpy(logits), dim=-1
    ).numpy()

    assert probs.shape == (1440, 6)
    assert np.isfinite(probs).all()

    v5_seed_probs.append(probs)

    print(f"Seed {seed}: completed")

    del model, tokenizer, logits, probs
    gc.collect()
    torch.cuda.empty_cache()

v5_test_probs = np.mean(
    np.stack(v5_seed_probs),
    axis=0
)

v5_test_labels = np.array(EMOTIONS)[
    v5_test_probs.argmax(axis=1)
]

# ----------------------------------------------------------
# 5. Verify reproduction of original V5
# ----------------------------------------------------------

print("\n========== V5 REPRODUCTION CHECK ==========")

old_v5 = pd.read_csv(
    "submission_v5.csv",
    dtype={"id": str}
)

assert old_v5["id"].is_unique
assert set(old_v5["id"]) == set(sample_ids)

original_labels = (
    old_v5.set_index("id")
    .loc[sample_ids, "label"]
    .to_numpy()
)

mismatch_mask = original_labels != v5_test_labels
mismatch_count = int(mismatch_mask.sum())

print("Original V5 rows:", len(original_labels))
print("Reproduced V5 mismatches:", mismatch_count)
print(
    "Agreement:",
    f"{1 - mismatch_count / 1440:.4%}"
)

if mismatch_count:
    mismatch_df = pd.DataFrame({
        "id": np.array(sample_ids)[mismatch_mask],
        "original": original_labels[mismatch_mask],
        "reproduced": v5_test_labels[mismatch_mask]
    })

    print(mismatch_df.head(10).to_string(index=False))

# Previously observed: one mismatch.
# Stop if the new reproduction deviates substantially.
if mismatch_count > 5:
    raise RuntimeError(
        "V5 reproduction differs too much. "
        "Do not generate the blend submission yet."
    )

# ----------------------------------------------------------
# 6. GoEmotions test prediction
# ----------------------------------------------------------

print("\n========== GOEMOTIONS TEST PREDICTION ==========")

go_tokenizer = AutoTokenizer.from_pretrained(
    GO_MODEL
)

go_model = AutoModelForSequenceClassification.from_pretrained(
    GO_MODEL
)

go_label_map = {
    str(name).lower(): int(idx)
    for idx, name in go_model.config.id2label.items()
}

assert all(
    emotion in go_label_map
    for emotion in EMOTIONS
)

go_indices = [
    go_label_map[emotion]
    for emotion in EMOTIONS
]

print("Selected labels:", go_indices)

go_logits = get_logits(
    go_model, go_tokenizer, test_texts
)

# Original model: 28-label sigmoid outputs
go_scores = torch.sigmoid(
    torch.from_numpy(go_logits)
).numpy()[:, go_indices]

assert go_scores.shape == (1440, 6)
assert np.isfinite(go_scores).all()

# Same normalization as validation experiment
go_normalized = go_scores / np.maximum(
    go_scores.sum(axis=1, keepdims=True),
    1e-12
)

del go_model, go_tokenizer, go_logits
gc.collect()
torch.cuda.empty_cache()

# ----------------------------------------------------------
# 7. Fixed 50/50 blending
# ----------------------------------------------------------

combined_probs = (
    0.5 * v5_test_probs
    + 0.5 * go_normalized
)

assert combined_probs.shape == (1440, 6)
assert np.isfinite(combined_probs).all()

combined_labels = np.array(EMOTIONS)[
    combined_probs.argmax(axis=1)
]

# ----------------------------------------------------------
# 8. Compare with original V5
# ----------------------------------------------------------

changed = int(
    np.sum(combined_labels != original_labels)
)

print("\n========== MODEL COMPARISON ==========")

print("V5 / GoEmotions blend:", "50% / 50%")
print("Total samples:", len(sample_ids))
print("Changed vs original V5:", changed)
print("Unchanged:", 1440 - changed)
print(
    "Agreement:",
    f"{1 - changed / 1440:.2%}"
)

distribution = pd.DataFrame({
    "Emotion": EMOTIONS,
    "Original V5": [
        int((original_labels == e).sum())
        for e in EMOTIONS
    ],
    "GoEmotions": [
        int((go_scores.argmax(axis=1) == i).sum())
        for i in range(6)
    ],
    "50/50 Blend": [
        int((combined_labels == e).sum())
        for e in EMOTIONS
    ]
})

print("\n========== LABEL DISTRIBUTION ==========")
print(distribution.to_string(index=False))

# ----------------------------------------------------------
# 9. Validate final submission
# ----------------------------------------------------------

submission = pd.DataFrame({
    "id": sample_ids,
    "label": combined_labels
})

assert submission.shape == (1440, 2)
assert list(submission.columns) == ["id", "label"]
assert submission["id"].is_unique
assert not submission.isna().any().any()
assert set(submission["label"]).issubset(set(EMOTIONS))

# ----------------------------------------------------------
# 10. Save results independently
# ----------------------------------------------------------

OUT_DIR.mkdir(parents=True, exist_ok=False)

np.savez_compressed(
    OUT_DIR / "test_probabilities.npz",
    ids=np.array(sample_ids),
    v5_probs=v5_test_probs,
    go_scores=go_scores,
    go_normalized=go_normalized,
    combined_probs=combined_probs
)

distribution.to_csv(
    OUT_DIR / "label_distribution.csv",
    index=False
)

summary = {
    "experiment": "V5 + GoEmotions 50/50",
    "v5_seeds": SEEDS,
    "go_model": GO_MODEL,
    "max_length": MAX_LENGTH,
    "blend_weight_v5": 0.5,
    "blend_weight_go": 0.5,
    "v5_reproduction_mismatches": mismatch_count,
    "changed_vs_original_v5": changed,
    "test_samples": 1440
}

with open(
    OUT_DIR / "experiment_summary.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(summary, f, indent=2)

submission.to_csv(
    SUBMISSION_PATH,
    index=False,
    encoding="utf-8"
)

print("\n========== COMPLETED ==========")
print("Submission:", SUBMISSION_PATH)
print("Shape:", submission.shape)
print("Diagnostics:", OUT_DIR)
print("No training performed.")
print("No Kaggle submission performed.")
```

```python

# ==========================================================
# Experiment 27: V5 + GoEmotions Weight Sensitivity
# Test only 40/60, 50/50, 60/40
# No training / No inference / No Kaggle submission
# ==========================================================

import numpy as np
import pandas as pd
from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    f1_score,
    accuracy_score,
    precision_recall_fscore_support
)

OUT_DIR = Path("v5_goemotions_weight_sensitivity")

if OUT_DIR.exists():
    raise FileExistsError(
        "Results already exist. Existing files preserved."
    )

EMOTIONS = [
    "anger", "disgust", "fear",
    "joy", "sadness", "surprise"
]

# ----------------------------------------------------------
# 1. Load original validation probabilities
# ----------------------------------------------------------

v5_dir = Path("distilbert_ensemble_validation")

v5_probs = np.mean([
    np.load(v5_dir / f"validation_probs_seed_{seed}.npy")
    for seed in [42, 123, 2026]
], axis=0).astype(np.float64)

blend_50_probs = np.load(
    "v5_goemotions_50_50_analysis/"
    "combined_validation_probs.npy"
).astype(np.float64)

assert v5_probs.shape == blend_50_probs.shape == (672, 6)

# Original:
# blend_50 = 0.5 * V5 + 0.5 * GoEmotions
# Therefore:
# GoEmotions = 2 * blend_50 - V5

go_probs = 2.0 * blend_50_probs - v5_probs

assert np.isfinite(go_probs).all()
assert go_probs.min() >= -1e-6
assert np.allclose(
    go_probs.sum(axis=1),
    1.0,
    atol=1e-5
)

# ----------------------------------------------------------
# 2. Restore original validation labels
# ----------------------------------------------------------

_, val_df_check = train_test_split(
    train_df,
    test_size=0.2,
    random_state=42,
    stratify=train_df["label"]
)

val_df_check = val_df_check.reset_index(drop=True)

if pd.api.types.is_numeric_dtype(val_df_check["label"]):
    y_true = val_df_check["label"].to_numpy(dtype=int)
else:
    label_map = {
        name: i for i, name in enumerate(EMOTIONS)
    }

    y_true = (
        val_df_check["label"]
        .astype(str)
        .str.lower()
        .str.strip()
        .map(label_map)
        .to_numpy(dtype=int)
    )

assert len(y_true) == 672
assert np.isin(y_true, range(6)).all()

def evaluate(pred):
    return (
        f1_score(
            y_true,
            pred,
            labels=range(6),
            average="macro",
            zero_division=0
        ),
        accuracy_score(y_true, pred)
    )

# Verify previously reported results
v5_pred = v5_probs.argmax(axis=1)
original_50_pred = blend_50_probs.argmax(axis=1)

assert abs(evaluate(v5_pred)[0] - 0.620562) < 1e-5
assert abs(evaluate(original_50_pred)[0] - 0.658051) < 1e-5

print("Original V5 verified.")
print("Original 50/50 blend verified.")

# ----------------------------------------------------------
# 3. Fixed three-weight comparison
# ----------------------------------------------------------

weights = [0.4, 0.5, 0.6]

records = []
predictions = {}
class_results = {"Emotion": EMOTIONS}

baseline_pred = original_50_pred
baseline_f1 = evaluate(baseline_pred)[0]

for weight_v5 in weights:

    weight_go = 1.0 - weight_v5

    combined_probs = (
        weight_v5 * v5_probs
        + weight_go * go_probs
    )

    pred = combined_probs.argmax(axis=1)

    name = f"V5 {int(weight_v5*100)}% + GO {int(weight_go*100)}%"

    f1, acc = evaluate(pred)

    predictions[name] = pred

    class_f1 = precision_recall_fscore_support(
        y_true,
        pred,
        labels=range(6),
        zero_division=0
    )[2]

    class_results[name] = class_f1

    baseline_correct = baseline_pred == y_true
    candidate_correct = pred == y_true

    records.append({
        "Model": name,
        "Macro F1": f1,
        "Accuracy": acc,
        "Delta vs 50/50": f1 - baseline_f1,
        "Changed vs 50/50": int(
            (pred != baseline_pred).sum()
        ),
        "Fixed vs 50/50": int(
            (~baseline_correct & candidate_correct).sum()
        ),
        "Broken vs 50/50": int(
            (baseline_correct & ~candidate_correct).sum()
        )
    })

# ----------------------------------------------------------
# 4. Print results
# ----------------------------------------------------------

results_df = pd.DataFrame(records)

class_df = pd.DataFrame(class_results)

print("\n========== WEIGHT COMPARISON ==========")
print(results_df.round(6).to_string(index=False))

print("\n========== PER-CLASS F1 ==========")
print(class_df.round(4).to_string(index=False))

# ----------------------------------------------------------
# 5. Save results
# ----------------------------------------------------------

OUT_DIR.mkdir(parents=True, exist_ok=False)

results_df.to_csv(
    OUT_DIR / "weight_comparison.csv",
    index=False
)

class_df.to_csv(
    OUT_DIR / "per_class_f1.csv",
    index=False
)

print("\n========== COMPLETED ==========")
print("Saved:", OUT_DIR)
print("No training performed.")
print("No Kaggle submission generated.")
```

```python

# ==========================================================
# Experiment 30: Dynamic Confidence-Based Ensemble
# Compare against V5 40% + GoEmotions 60%
# No training / No inference / No Kaggle submission
# ==========================================================

import numpy as np
import pandas as pd
from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    f1_score,
    accuracy_score,
    precision_recall_fscore_support
)

OUT_DIR = Path("v5_goemotions_dynamic_weight")

if OUT_DIR.exists():
    raise FileExistsError(
        "Experiment already exists. Previous results preserved."
    )

EMOTIONS = [
    "anger", "disgust", "fear",
    "joy", "sadness", "surprise"
]

# ----------------------------------------------------------
# 1. Load saved validation probabilities
# ----------------------------------------------------------

v5_probs = np.mean([
    np.load(
        f"distilbert_ensemble_validation/"
        f"validation_probs_seed_{seed}.npy"
    )
    for seed in [42, 123, 2026]
], axis=0).astype(np.float64)

blend_50 = np.load(
    "v5_goemotions_50_50_analysis/"
    "combined_validation_probs.npy"
).astype(np.float64)

assert v5_probs.shape == blend_50.shape == (672, 6)

# Recover normalized GoEmotions probabilities
go_probs = 2.0 * blend_50 - v5_probs

assert np.isfinite(go_probs).all()
assert go_probs.min() >= -1e-6

go_probs = np.maximum(go_probs, 1e-12)
go_probs /= go_probs.sum(axis=1, keepdims=True)

# ----------------------------------------------------------
# 2. Restore original validation labels
# ----------------------------------------------------------

_, val_df = train_test_split(
    train_df,
    test_size=0.2,
    random_state=42,
    stratify=train_df["label"]
)

mapping = {
    emotion: i
    for i, emotion in enumerate(EMOTIONS)
}

if pd.api.types.is_numeric_dtype(val_df["label"]):
    y_true = val_df["label"].to_numpy(dtype=int)
else:
    y_true = (
        val_df["label"]
        .astype(str)
        .str.lower()
        .str.strip()
        .map(mapping)
        .to_numpy(dtype=int)
    )

assert len(y_true) == 672
assert np.isin(y_true, range(6)).all()

def evaluate(pred):
    return (
        f1_score(
            y_true, pred,
            labels=range(6),
            average="macro",
            zero_division=0
        ),
        accuracy_score(y_true, pred)
    )

# ----------------------------------------------------------
# 3. Fixed 40/60 baseline
# ----------------------------------------------------------

baseline_probs = (
    0.4 * v5_probs
    + 0.6 * go_probs
)

baseline_pred = baseline_probs.argmax(axis=1)

baseline_f1, baseline_acc = evaluate(baseline_pred)

assert abs(baseline_f1 - 0.667754) < 1e-5

print("40/60 Baseline verified:", round(baseline_f1, 6))

# ----------------------------------------------------------
# 4. Calculate confidence via top-2 margin
# ----------------------------------------------------------

def relative_margin(probs):
    top2 = np.sort(probs, axis=1)[:, -2:]

    return (
        (top2[:, 1] - top2[:, 0])
        / (top2[:, 1] + top2[:, 0] + 1e-12)
    )

v5_margin = relative_margin(v5_probs)
go_margin = relative_margin(go_probs)

# Fixed, predefined dynamic-weight rule:
# More confidence in GO -> increase GO weight
# More confidence in V5 -> decrease GO weight

relative_confidence = (
    (go_margin - v5_margin)
    / (go_margin + v5_margin + 1e-12)
)

go_weights = np.clip(
    0.60 + 0.15 * relative_confidence,
    0.45,
    0.75
)

v5_weights = 1.0 - go_weights

# ----------------------------------------------------------
# 5. Dynamic ensemble prediction
# ----------------------------------------------------------

dynamic_probs = (
    v5_weights[:, None] * v5_probs
    + go_weights[:, None] * go_probs
)

dynamic_pred = dynamic_probs.argmax(axis=1)

dynamic_f1, dynamic_acc = evaluate(dynamic_pred)

# ----------------------------------------------------------
# 6. Comparison
# ----------------------------------------------------------

baseline_correct = baseline_pred == y_true
dynamic_correct = dynamic_pred == y_true

fixed = int(
    (~baseline_correct & dynamic_correct).sum()
)

broken = int(
    (baseline_correct & ~dynamic_correct).sum()
)

comparison = pd.DataFrame([
    {
        "Model": "Fixed 40/60",
        "Macro F1": baseline_f1,
        "Accuracy": baseline_acc
    },
    {
        "Model": "Dynamic Weight",
        "Macro F1": dynamic_f1,
        "Accuracy": dynamic_acc
    }
])

print("\n========== MODEL COMPARISON ==========")
print(comparison.round(6).to_string(index=False))

print(
    "\nDelta Macro F1:",
    f"{dynamic_f1 - baseline_f1:+.6f}"
)

print("\n========== DYNAMIC WEIGHT STATISTICS ==========")
print("GO Weight Minimum:", round(go_weights.min(), 4))
print("GO Weight Mean:", round(go_weights.mean(), 4))
print("GO Weight Median:", round(np.median(go_weights), 4))
print("GO Weight Maximum:", round(go_weights.max(), 4))

print("\n========== PREDICTION CHANGES ==========")
print(
    "Changed:",
    int((baseline_pred != dynamic_pred).sum())
)
print("Wrong -> Correct:", fixed)
print("Correct -> Wrong:", broken)
print("Net Correct:", fixed - broken)

# ----------------------------------------------------------
# 7. Per-class F1
# ----------------------------------------------------------

def class_f1(pred):
    return precision_recall_fscore_support(
        y_true,
        pred,
        labels=range(6),
        zero_division=0
    )[2]

baseline_class = class_f1(baseline_pred)
dynamic_class = class_f1(dynamic_pred)

per_class = pd.DataFrame({
    "Emotion": EMOTIONS,
    "Fixed 40/60 F1": baseline_class,
    "Dynamic F1": dynamic_class,
    "Difference": dynamic_class - baseline_class
})

print("\n========== PER-CLASS F1 ==========")
print(per_class.round(4).to_string(index=False))

# ----------------------------------------------------------
# 8. Save results
# ----------------------------------------------------------

OUT_DIR.mkdir(parents=True, exist_ok=False)

comparison.to_csv(
    OUT_DIR / "model_comparison.csv",
    index=False
)

per_class.to_csv(
    OUT_DIR / "per_class_f1.csv",
    index=False
)

np.save(
    OUT_DIR / "dynamic_validation_probs.npy",
    dynamic_probs
)

print("\n========== COMPLETED ==========")
print("Saved:", OUT_DIR)
print("Original 40/60 preserved.")
print("No training or Kaggle submission.")
```

```python

# ==========================================================
# Experiment 31: Fine-tune GoEmotions
# Preserve original 28-label classification head
# Compare against fixed 40/60 validation baseline
# No Kaggle submission
# ==========================================================

import gc
import random
import numpy as np
import pandas as pd
import torch
import torch.nn.functional as F

from pathlib import Path
from torch.utils.data import Dataset, DataLoader
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    DataCollatorWithPadding,
    get_linear_schedule_with_warmup
)
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score, accuracy_score

# ----------------------------------------------------------
# 1. Settings
# ----------------------------------------------------------

MODEL_NAME = "SamLowe/roberta-base-go_emotions"

OUT_DIR = Path("models/goemotions_finetune_seed123")
MODEL_DIR = OUT_DIR / "final_model"

if OUT_DIR.exists():
    raise FileExistsError(
        "Experiment directory exists. Existing files preserved."
    )

EMOTIONS = [
    "anger", "disgust", "fear",
    "joy", "sadness", "surprise"
]

SEED = 123
EPOCHS = 2
LR = 1e-5
MAX_LENGTH = 128
BATCH_SIZE = 4
GRAD_ACCUM = 2

assert torch.cuda.is_available(), "CUDA unavailable."

device = torch.device("cuda")

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

print("GPU:", torch.cuda.get_device_name(0))

# ----------------------------------------------------------
# 2. Original train/validation split
# ----------------------------------------------------------

train_part, val_part = train_test_split(
    train_df,
    test_size=0.2,
    random_state=42,
    stratify=train_df["label"]
)

train_part = train_part.reset_index(drop=True)
val_part = val_part.reset_index(drop=True)

label_map = {
    name: i for i, name in enumerate(EMOTIONS)
}

def encode_labels(series):
    if pd.api.types.is_numeric_dtype(series):
        return series.to_numpy(dtype=int)

    return (
        series.astype(str)
        .str.lower()
        .str.strip()
        .map(label_map)
        .to_numpy(dtype=int)
    )

train_y = encode_labels(train_part["label"])
val_y = encode_labels(val_part["label"])

assert len(train_part) == 2688
assert len(val_part) == 672
assert np.isin(train_y, range(6)).all()
assert np.isin(val_y, range(6)).all()

# ----------------------------------------------------------
# 3. Load original 40/60 baseline
# ----------------------------------------------------------

v5_probs = np.mean([
    np.load(
        "distilbert_ensemble_validation/"
        f"validation_probs_seed_{s}.npy"
    )
    for s in [42, 123, 2026]
], axis=0).astype(np.float64)

blend50 = np.load(
    "v5_goemotions_50_50_analysis/"
    "combined_validation_probs.npy"
).astype(np.float64)

go_original_probs = 2.0 * blend50 - v5_probs

baseline_probs = (
    0.4 * v5_probs
    + 0.6 * go_original_probs
)

def macro_f1(pred):
    return f1_score(
        val_y, pred,
        labels=range(6),
        average="macro",
        zero_division=0
    )

baseline_f1 = macro_f1(
    baseline_probs.argmax(axis=1)
)

assert abs(baseline_f1 - 0.667754) < 1e-5

print("Original 40/60 baseline:", round(baseline_f1, 6))

# ----------------------------------------------------------
# 4. Load emotion-pretrained model
# ----------------------------------------------------------

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME
)

model_labels = {
    str(name).lower(): int(idx)
    for idx, name in model.config.id2label.items()
}

assert all(e in model_labels for e in EMOTIONS)

selected_indices = torch.tensor(
    [model_labels[e] for e in EMOTIONS],
    dtype=torch.long,
    device=device
)

print("Selected emotion indices:", selected_indices.tolist())
print("Original output classes:", model.config.num_labels)

assert model.config.num_labels == 28

model.to(device)

# ----------------------------------------------------------
# 5. Dataset and dataloaders
# ----------------------------------------------------------

class EmotionDataset(Dataset):

    def __init__(self, texts, labels):
        self.texts = texts
        self.labels = labels

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        item = tokenizer(
            self.texts[idx],
            truncation=True,
            max_length=MAX_LENGTH
        )
        item["labels"] = int(self.labels[idx])
        return item

train_ds = EmotionDataset(
    train_part["text"].fillna("").astype(str).tolist(),
    train_y
)

val_ds = EmotionDataset(
    val_part["text"].fillna("").astype(str).tolist(),
    val_y
)

collator = DataCollatorWithPadding(
    tokenizer=tokenizer
)

generator = torch.Generator()
generator.manual_seed(SEED)

train_loader = DataLoader(
    train_ds,
    batch_size=BATCH_SIZE,
    shuffle=True,
    generator=generator,
    collate_fn=collator
)

val_loader = DataLoader(
    val_ds,
    batch_size=16,
    shuffle=False,
    collate_fn=collator
)

# ----------------------------------------------------------
# 6. Optimizer and scheduler
# ----------------------------------------------------------

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=LR,
    weight_decay=0.01
)

total_steps = (
    (len(train_loader) + GRAD_ACCUM - 1)
    // GRAD_ACCUM
) * EPOCHS

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=0,
    num_training_steps=total_steps
)

scaler = torch.amp.GradScaler("cuda")

# ----------------------------------------------------------
# 7. Validation function
# ----------------------------------------------------------

def evaluate_current_model():

    model.eval()

    all_scores = []

    with torch.inference_mode():

        for batch in val_loader:

            batch.pop("labels")

            batch = {
                k: v.to(device)
                for k, v in batch.items()
            }

            with torch.autocast(
                device_type="cuda",
                dtype=torch.float16
            ):
                logits = model(**batch).logits

            selected = logits.index_select(
                dim=1,
                index=selected_indices
            )

            scores = torch.sigmoid(
                selected.float()
            )

            all_scores.append(
                scores.cpu().numpy()
            )

    scores = np.concatenate(all_scores, axis=0)

    # Same sigmoid normalization used previously
    normalized = scores / np.maximum(
        scores.sum(axis=1, keepdims=True),
        1e-12
    )

    single_pred = normalized.argmax(axis=1)

    combined_probs = (
        0.4 * v5_probs
        + 0.6 * normalized
    )

    combined_pred = combined_probs.argmax(axis=1)

    return {
        "GO F1": macro_f1(single_pred),
        "GO Accuracy": accuracy_score(val_y, single_pred),
        "Blend 40/60 F1": macro_f1(combined_pred),
        "Blend Accuracy": accuracy_score(val_y, combined_pred)
    }

# ----------------------------------------------------------
# 8. Fine-tuning
# ----------------------------------------------------------

history = []

print("\n========== GOEMOTIONS FINE-TUNING ==========")

for epoch in range(EPOCHS):

    model.train()
    optimizer.zero_grad(set_to_none=True)

    total_loss = 0.0

    for step, batch in enumerate(train_loader):

        labels = batch.pop("labels").to(device)

        batch = {
            k: v.to(device)
            for k, v in batch.items()
        }

        with torch.autocast(
            device_type="cuda",
            dtype=torch.float16
        ):

            logits = model(**batch).logits

            selected_logits = logits.index_select(
                dim=1,
                index=selected_indices
            )

            loss = F.cross_entropy(
                selected_logits.float(),
                labels
            )

        # Account for a possibly shorter final accumulation group
        group_start = (step // GRAD_ACCUM) * GRAD_ACCUM
        group_size = min(
            GRAD_ACCUM,
            len(train_loader) - group_start
        )

        scaler.scale(loss / group_size).backward()

        total_loss += loss.item()

        update_now = (
            (step + 1) % GRAD_ACCUM == 0
            or (step + 1) == len(train_loader)
        )

        if update_now:

            scaler.unscale_(optimizer)

            torch.nn.utils.clip_grad_norm_(
                model.parameters(),
                max_norm=1.0
            )

            scaler.step(optimizer)
            scaler.update()

            scheduler.step()

            optimizer.zero_grad(set_to_none=True)

    metrics = evaluate_current_model()

    row = {
        "Epoch": epoch + 1,
        "Train Loss": total_loss / len(train_loader),
        **metrics
    }

    history.append(row)

    print(f"\nEpoch {epoch + 1}/{EPOCHS}")
    print("Train Loss:", round(row["Train Loss"], 6))
    print("GoEmotions F1:", round(row["GO F1"], 6))
    print("Blend 40/60 F1:", round(row["Blend 40/60 F1"], 6))
    print("Blend Accuracy:", round(row["Blend Accuracy"], 6))

# ----------------------------------------------------------
# 9. Save final 2-epoch model
# ----------------------------------------------------------

OUT_DIR.mkdir(parents=True, exist_ok=False)

model.save_pretrained(str(MODEL_DIR))
tokenizer.save_pretrained(str(MODEL_DIR))

history_df = pd.DataFrame(history)

history_df.to_csv(
    OUT_DIR / "training_history.csv",
    index=False
)

print("\n========== FINAL COMPARISON ==========")

print("Original GO F1: 0.627871")
print("Original 40/60 F1:", round(baseline_f1, 6))

print(
    history_df[
        ["Epoch", "GO F1", "Blend 40/60 F1", "Blend Accuracy"]
    ].round(6).to_string(index=False)
)

print("\n========== COMPLETED ==========")
print("Model saved:", MODEL_DIR)
print("Original GoEmotions unchanged.")
print("Original V5 unchanged.")
print("Original 40/60 unchanged.")
print("No Kaggle submission generated.")

# Free GPU memory
del model, optimizer, scheduler, scaler
gc.collect()
torch.cuda.empty_cache()
```

```python

# ==========================================================
# Experiment 34: Disgust Decision Bias
# Baseline: V5 40% + GoEmotions 60%
# No training / No inference / No Kaggle submission
# ==========================================================

import numpy as np
import pandas as pd

from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    f1_score,
    accuracy_score,
    precision_recall_fscore_support
)

OUT_DIR = Path("v5_goemotions_disgust_bias_analysis")

if OUT_DIR.exists():
    raise FileExistsError(
        "Experiment already exists. Previous results preserved."
    )

EMOTIONS = [
    "anger", "disgust", "fear",
    "joy", "sadness", "surprise"
]

DISGUST_ID = 1

# ----------------------------------------------------------
# 1. Restore saved model predictions
# ----------------------------------------------------------

v5_probs = np.mean([
    np.load(
        f"distilbert_ensemble_validation/"
        f"validation_probs_seed_{s}.npy"
    )
    for s in [42, 123, 2026]
], axis=0).astype(np.float64)

blend50 = np.load(
    "v5_goemotions_50_50_analysis/"
    "combined_validation_probs.npy"
).astype(np.float64)

assert v5_probs.shape == blend50.shape == (672, 6)

go_probs = 2.0 * blend50 - v5_probs

baseline_probs = (
    0.4 * v5_probs + 0.6 * go_probs
)

assert np.isfinite(baseline_probs).all()

# ----------------------------------------------------------
# 2. Restore original validation labels
# ----------------------------------------------------------

_, val_df = train_test_split(
    train_df,
    test_size=0.2,
    random_state=42,
    stratify=train_df["label"]
)

label_map = {
    name: i for i, name in enumerate(EMOTIONS)
}

if pd.api.types.is_numeric_dtype(val_df["label"]):
    y_true = val_df["label"].to_numpy(dtype=int)
else:
    y_true = (
        val_df["label"]
        .astype(str)
        .str.lower()
        .str.strip()
        .map(label_map)
        .to_numpy(dtype=int)
    )

assert len(y_true) == 672
assert np.isin(y_true, range(6)).all()

baseline_pred = baseline_probs.argmax(axis=1)

def macro_f1(pred):
    return f1_score(
        y_true,
        pred,
        labels=range(6),
        average="macro",
        zero_division=0
    )

baseline_f1 = macro_f1(baseline_pred)

assert abs(baseline_f1 - 0.667754) < 1e-5

print(
    "Original 40/60 baseline verified:",
    round(baseline_f1, 6)
)

# ----------------------------------------------------------
# 3. Check Disgust score margins
# ----------------------------------------------------------

disgust_scores = baseline_probs[:, DISGUST_ID]

other_scores = np.delete(
    baseline_probs,
    DISGUST_ID,
    axis=1
).max(axis=1)

missed_disgust = (
    (y_true == DISGUST_ID)
    & (baseline_pred != DISGUST_ID)
)

# Minimum multiplier needed to beat the best other class
required_multiplier = (
    other_scores[missed_disgust]
    / np.maximum(
        disgust_scores[missed_disgust],
        1e-12
    )
)

print("\n========== MISSED DISGUST MARGINS ==========")
print("Missed Disgust:", int(missed_disgust.sum()))

for factor in [1.10, 1.25]:
    potentially_recovered = int(
        (required_multiplier < factor).sum()
    )
    print(
        f"Recoverable with multiplier {factor:.2f}:",
        potentially_recovered
    )

# ----------------------------------------------------------
# 4. Fixed multiplier experiment
# ----------------------------------------------------------

multipliers = [1.00, 1.10, 1.25]

results = []
per_class = {"Emotion": EMOTIONS}

base_correct = baseline_pred == y_true

for factor in multipliers:

    # Modify decision scores only.
    # These are NOT calibrated probabilities.
    scores = baseline_probs.copy()

    scores[:, DISGUST_ID] *= factor

    pred = scores.argmax(axis=1)

    precision, recall, f1_classes, _ = (
        precision_recall_fscore_support(
            y_true,
            pred,
            labels=range(6),
            zero_division=0
        )
    )

    candidate_correct = pred == y_true

    fixed = int(
        (~base_correct & candidate_correct).sum()
    )

    broken = int(
        (base_correct & ~candidate_correct).sum()
    )

    recovered_disgust = int(
        (
            (y_true == DISGUST_ID)
            & (baseline_pred != DISGUST_ID)
            & (pred == DISGUST_ID)
        ).sum()
    )

    new_false_disgust = int(
        (
            (y_true != DISGUST_ID)
            & (baseline_pred != DISGUST_ID)
            & (pred == DISGUST_ID)
        ).sum()
    )

    results.append({
        "Disgust Multiplier": factor,
        "Macro F1": macro_f1(pred),
        "Accuracy": accuracy_score(y_true, pred),
        "Disgust Precision": precision[DISGUST_ID],
        "Disgust Recall": recall[DISGUST_ID],
        "Disgust F1": f1_classes[DISGUST_ID],
        "Changed": int((pred != baseline_pred).sum()),
        "Fixed": fixed,
        "Broken": broken,
        "Recovered Disgust": recovered_disgust,
        "New False Disgust": new_false_disgust
    })

    per_class[f"x{factor:.2f}"] = f1_classes

# ----------------------------------------------------------
# 5. Print results
# ----------------------------------------------------------

results_df = pd.DataFrame(results)
per_class_df = pd.DataFrame(per_class)

print("\n========== DISGUST BIAS COMPARISON ==========")
print(
    results_df.round(6).to_string(index=False)
)

print("\n========== PER-CLASS F1 ==========")
print(
    per_class_df.round(4).to_string(index=False)
)

# ----------------------------------------------------------
# 6. Save analysis
# ----------------------------------------------------------

OUT_DIR.mkdir(parents=True, exist_ok=False)

results_df.to_csv(
    OUT_DIR / "disgust_bias_comparison.csv",
    index=False
)

per_class_df.to_csv(
    OUT_DIR / "per_class_f1.csv",
    index=False
)

print("\n========== COMPLETED ==========")
print("Saved:", OUT_DIR)
print("Original 40/60 preserved.")
print("No training or Kaggle submission.")
```

```python

# ==========================================================
# Experiment 35: GoEmotions 28-label Disgust Diagnostic
# Investigate auxiliary emotion signals
# No training / No Kaggle submission
# ==========================================================

import gc
import numpy as np
import pandas as pd
import torch

from pathlib import Path
from collections import Counter
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_auc_score
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification
)

MODEL_NAME = "SamLowe/roberta-base-go_emotions"

DIAGNOSTIC_PATH = Path(
    "v5_goemotions_40_60_error_analysis/"
    "validation_predictions.csv"
)

OUT_DIR = Path(
    "goemotions_28label_disgust_diagnostic"
)

if OUT_DIR.exists():
    raise FileExistsError(
        "Output directory exists. Existing results preserved."
    )

assert DIAGNOSTIC_PATH.exists()
assert torch.cuda.is_available()

device = torch.device("cuda")

# ----------------------------------------------------------
# 1. Restore validation sample order
# ----------------------------------------------------------

_, val_df = train_test_split(
    train_df,
    test_size=0.2,
    random_state=42,
    stratify=train_df["label"]
)

val_df = val_df.reset_index(drop=True)

diagnostics = pd.read_csv(
    DIAGNOSTIC_PATH,
    dtype={"post_id": str}
)

assert len(val_df) == len(diagnostics) == 672

assert np.array_equal(
    val_df["post_id"].astype(str).to_numpy(),
    diagnostics["post_id"].to_numpy()
), "Validation sample alignment mismatch."

texts = (
    val_df["text"]
    .fillna("")
    .astype(str)
    .tolist()
)

actual = diagnostics["actual"].astype(str).to_numpy()
blend = diagnostics["blend_prediction"].astype(str).to_numpy()
v5 = diagnostics["V5_prediction"].astype(str).to_numpy()
go = diagnostics["GO_prediction"].astype(str).to_numpy()

actual_disgust = actual == "disgust"

missed_disgust = (
    actual_disgust &
    (blend != "disgust")
)

correct_disgust = (
    actual_disgust &
    (blend == "disgust")
)

both_wrong = (
    missed_disgust &
    (v5 != "disgust") &
    (go != "disgust")
)

non_disgust = ~actual_disgust

print("========== VALIDATION GROUPS ==========")
print("Actual Disgust:", int(actual_disgust.sum()))
print("Missed Disgust:", int(missed_disgust.sum()))
print("Correct Disgust:", int(correct_disgust.sum()))
print("Both models wrong:", int(both_wrong.sum()))

assert missed_disgust.sum() == 56
assert correct_disgust.sum() == 35
assert both_wrong.sum() == 43

# ----------------------------------------------------------
# 2. Load original GoEmotions model
# ----------------------------------------------------------

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME
)

assert model.config.num_labels == 28

label_names = [
    str(model.config.id2label[i]).lower()
    for i in range(28)
]

model.to(device)
model.eval()

# ----------------------------------------------------------
# 3. Obtain ALL 28 emotion scores
# ----------------------------------------------------------

all_scores = []

with torch.inference_mode():

    for start in range(0, len(texts), 16):

        batch = texts[start:start + 16]

        inputs = tokenizer(
            batch,
            padding=True,
            truncation=True,
            max_length=128,
            return_tensors="pt"
        ).to(device)

        logits = model(**inputs).logits

        scores = torch.sigmoid(
            logits.float()
        )

        all_scores.append(
            scores.cpu().numpy()
        )

scores_28 = np.concatenate(all_scores, axis=0)

assert scores_28.shape == (672, 28)
assert np.isfinite(scores_28).all()

# ----------------------------------------------------------
# 4. Compare emotion scores across groups
# ----------------------------------------------------------

rows = []

for i, emotion in enumerate(label_names):

    column = scores_28[:, i]

    rows.append({
        "Emotion": emotion,
        "Mean Missed Disgust": column[missed_disgust].mean(),
        "Mean Correct Disgust": column[correct_disgust].mean(),
        "Mean Non-Disgust": column[non_disgust].mean(),
        "AUC for Disgust": roc_auc_score(
            actual_disgust.astype(int),
            column
        )
    })

report = pd.DataFrame(rows)

print("\n========== TOP LABELS BY DISGUST AUC ==========")

print(
    report
    .sort_values("AUC for Disgust", ascending=False)
    .head(12)
    .round(4)
    .to_string(index=False)
)

# ----------------------------------------------------------
# 5. Inspect auxiliary emotion labels
# ----------------------------------------------------------

interesting = [
    "disgust",
    "annoyance",
    "disapproval",
    "disappointment",
    "anger",
    "confusion",
    "neutral"
]

print("\n========== RELEVANT EMOTION SIGNALS ==========")

print(
    report[
        report["Emotion"].isin(interesting)
    ]
    .round(4)
    .to_string(index=False)
)

# ----------------------------------------------------------
# 6. Top-3 labels among 43 jointly missed cases
# ----------------------------------------------------------

top3_count = Counter()

for row in scores_28[both_wrong]:
    indices = np.argsort(row)[-3:]

    for idx in indices:
        top3_count[label_names[idx]] += 1

print("\n========== BOTH-WRONG TOP-3 LABEL FREQUENCY ==========")

for emotion, count in top3_count.most_common(12):
    print(f"{emotion:20s} {count}")

# ----------------------------------------------------------
# 7. Save results
# ----------------------------------------------------------

OUT_DIR.mkdir(parents=True, exist_ok=False)

report.to_csv(
    OUT_DIR / "emotion_signal_comparison.csv",
    index=False
)

np.save(
    OUT_DIR / "validation_28label_scores.npy",
    scores_28
)

print("\n========== COMPLETED ==========")
print("Saved:", OUT_DIR)
print("No model modifications.")
print("No training or Kaggle submission.")

del model, tokenizer
gc.collect()
torch.cuda.empty_cache()
```

```python

# ==========================================================
# Experiment 39 - Step 2
# Independent Holdout: DistilBERT 3 Seeds + GoEmotions
# Fixed comparisons: 50/50 vs 40/60
# No Kaggle submission
# ==========================================================

import gc
import random
import numpy as np
import pandas as pd
import torch

from pathlib import Path
from torch.utils.data import Dataset

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer,
    DataCollatorWithPadding,
    set_seed
)

from sklearn.metrics import f1_score, accuracy_score

# ----------------------------------------------------------
# 1. Settings
# ----------------------------------------------------------

SPLIT_DIR = Path("independent_holdout_2027")
OUT_DIR = SPLIT_DIR / "distilbert_3seed_comparison"

if OUT_DIR.exists():
    raise FileExistsError(
        "Experiment already exists. Existing results preserved."
    )

DISTILBERT_NAME = "distilbert/distilbert-base-uncased"
GO_MODEL_NAME = "SamLowe/roberta-base-go_emotions"

SEEDS = [42, 123, 2026]
EMOTIONS = [
    "anger", "disgust", "fear",
    "joy", "sadness", "surprise"
]

MAX_LENGTH = 128
BATCH_SIZE = 8

assert torch.cuda.is_available(), "CUDA unavailable."
device = torch.device("cuda")

# ----------------------------------------------------------
# 2. Load saved independent holdout indices
# ----------------------------------------------------------

train_idx = np.load(SPLIT_DIR / "train_indices.npy")
val_idx = np.load(SPLIT_DIR / "validation_indices.npy")
old_val_idx = np.load(
    SPLIT_DIR / "old_validation_indices.npy"
)

assert len(train_idx) == 2150
assert len(val_idx) == 538
assert len(set(train_idx) & set(val_idx)) == 0
assert len(set(val_idx) & set(old_val_idx)) == 0

new_train = train_df.iloc[train_idx].reset_index(drop=True)
new_val = train_df.iloc[val_idx].reset_index(drop=True)

label_map = {e: i for i, e in enumerate(EMOTIONS)}
id2label = {i: e for i, e in enumerate(EMOTIONS)}

def encode_labels(series):
    if pd.api.types.is_numeric_dtype(series):
        return series.to_numpy(dtype=int)

    return (
        series.astype(str)
        .str.lower()
        .str.strip()
        .map(label_map)
        .to_numpy(dtype=int)
    )

train_y = encode_labels(new_train["label"])
val_y = encode_labels(new_val["label"])

assert np.isin(train_y, range(6)).all()
assert np.isin(val_y, range(6)).all()

train_texts = (
    new_train["text"].fillna("").astype(str).tolist()
)

val_texts = (
    new_val["text"].fillna("").astype(str).tolist()
)

def evaluate(pred):
    return (
        f1_score(
            val_y, pred,
            labels=range(6),
            average="macro",
            zero_division=0
        ),
        accuracy_score(val_y, pred)
    )

print("GPU:", torch.cuda.get_device_name(0))
print("New Training:", len(train_texts))
print("New Holdout:", len(val_texts))

# ----------------------------------------------------------
# 3. Prepare DistilBERT datasets
# ----------------------------------------------------------

tokenizer = AutoTokenizer.from_pretrained(
    DISTILBERT_NAME
)

class EmotionDataset(Dataset):

    def __init__(self, texts, labels, tokenizer):
        self.encodings = tokenizer(
            texts,
            truncation=True,
            max_length=MAX_LENGTH
        )
        self.labels = labels

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        item = {
            k: v[idx]
            for k, v in self.encodings.items()
        }
        item["labels"] = int(self.labels[idx])
        return item

train_ds = EmotionDataset(
    train_texts, train_y, tokenizer
)

val_ds = EmotionDataset(
    val_texts, val_y, tokenizer
)

collator = DataCollatorWithPadding(
    tokenizer=tokenizer
)

# ----------------------------------------------------------
# 4. Retrain the original three DistilBERT seeds
# ----------------------------------------------------------

OUT_DIR.mkdir(parents=True, exist_ok=False)

all_seed_probs = []
seed_results = []

print("\n========== DISTILBERT HOLDOUT TRAINING ==========")

for seed in SEEDS:

    print(f"\n========== SEED {seed} ==========")

    set_seed(seed)

    model = AutoModelForSequenceClassification.from_pretrained(
        DISTILBERT_NAME,
        num_labels=6,
        id2label=id2label,
        label2id=label_map
    )

    args = TrainingArguments(
        output_dir=str(OUT_DIR / f"trainer_seed_{seed}"),
        learning_rate=2e-5,
        num_train_epochs=3,
        per_device_train_batch_size=BATCH_SIZE,
        per_device_eval_batch_size=BATCH_SIZE,
        gradient_accumulation_steps=1,
        weight_decay=0.01,
        lr_scheduler_type="linear",
        warmup_steps=0,
        fp16=True,
        eval_strategy="no",
        save_strategy="no",
        logging_strategy="epoch",
        seed=seed,
        data_seed=seed,
        report_to="none"
    )

    trainer = Trainer(
        model=model,
        args=args,
        train_dataset=train_ds,
        processing_class=tokenizer,
        data_collator=collator
    )

    trainer.train()

    prediction = trainer.predict(val_ds)
    logits = prediction.predictions

    logits = logits - logits.max(
        axis=1, keepdims=True
    )

    probs = np.exp(logits)
    probs /= probs.sum(axis=1, keepdims=True)

    assert probs.shape == (538, 6)
    assert np.isfinite(probs).all()

    pred = probs.argmax(axis=1)
    f1, acc = evaluate(pred)

    print(f"Seed {seed} Holdout F1: {f1:.6f}")
    print(f"Seed {seed} Accuracy: {acc:.6f}")

    all_seed_probs.append(probs)

    seed_results.append({
        "Seed": seed,
        "Macro F1": f1,
        "Accuracy": acc
    })

    # Save each trained model and its predictions
    model_dir = OUT_DIR / f"seed_{seed}"

    trainer.save_model(str(model_dir))
    tokenizer.save_pretrained(str(model_dir))

    np.save(
        OUT_DIR / f"val_probs_seed_{seed}.npy",
        probs
    )

    pd.DataFrame(
        trainer.state.log_history
    ).to_csv(
        OUT_DIR / f"train_log_seed_{seed}.csv",
        index=False
    )

    del trainer, model, args, prediction, logits
    gc.collect()
    torch.cuda.empty_cache()

# Average all three DistilBERT predictions
v5_new_probs = np.mean(
    np.stack(all_seed_probs),
    axis=0
)

# ----------------------------------------------------------
# 5. Original GoEmotions: zero-shot on NEW holdout
# ----------------------------------------------------------

print("\n========== GOEMOTIONS HOLDOUT PREDICTION ==========")

go_tokenizer = AutoTokenizer.from_pretrained(
    GO_MODEL_NAME
)

go_model = AutoModelForSequenceClassification.from_pretrained(
    GO_MODEL_NAME
)

go_labels = {
    str(name).lower(): int(idx)
    for idx, name in go_model.config.id2label.items()
}

go_indices = [
    go_labels[e] for e in EMOTIONS
]

assert go_model.config.num_labels == 28

go_model.to(device)
go_model.eval()

all_go_scores = []

with torch.inference_mode():

    for start in range(0, len(val_texts), 16):

        batch = val_texts[start:start + 16]

        inputs = go_tokenizer(
            batch,
            padding=True,
            truncation=True,
            max_length=128,
            return_tensors="pt"
        ).to(device)

        logits = go_model(**inputs).logits

        scores = torch.sigmoid(
            logits.float()
        )[:, go_indices]

        all_go_scores.append(
            scores.cpu().numpy()
        )

go_scores = np.concatenate(
    all_go_scores,
    axis=0
)

go_probs = go_scores / np.maximum(
    go_scores.sum(axis=1, keepdims=True),
    1e-12
)

assert go_probs.shape == (538, 6)

del go_model, go_tokenizer
gc.collect()
torch.cuda.empty_cache()

# ----------------------------------------------------------
# 6. Fixed comparison: 50/50 vs 40/60
# ----------------------------------------------------------

experiments = {
    "DistilBERT 3 Seeds": v5_new_probs,
    "GoEmotions Zero-shot": go_probs,
    "V5 50% + GO 50%": (
        0.5 * v5_new_probs + 0.5 * go_probs
    ),
    "V5 40% + GO 60%": (
        0.4 * v5_new_probs + 0.6 * go_probs
    )
}

results = []

for name, probs in experiments.items():

    pred = probs.argmax(axis=1)
    f1, acc = evaluate(pred)

    results.append({
        "Model": name,
        "Macro F1": f1,
        "Accuracy": acc
    })

results_df = pd.DataFrame(results)

pred_50 = experiments[
    "V5 50% + GO 50%"
].argmax(axis=1)

pred_40 = experiments[
    "V5 40% + GO 60%"
].argmax(axis=1)

correct_50 = pred_50 == val_y
correct_40 = pred_40 == val_y

f1_50 = evaluate(pred_50)[0]
f1_40 = evaluate(pred_40)[0]

# ----------------------------------------------------------
# 7. Print results
# ----------------------------------------------------------

print("\n========== INDIVIDUAL SEED RESULTS ==========")
print(
    pd.DataFrame(seed_results)
    .round(6)
    .to_string(index=False)
)

print("\n========== NEW HOLDOUT COMPARISON ==========")
print(
    results_df
    .round(6)
    .to_string(index=False)
)

print("\n========== 40/60 VS 50/50 ==========")
print("Delta Macro F1:", round(f1_40 - f1_50, 6))
print("Changed Predictions:", int((pred_40 != pred_50).sum()))
print(
    "50/50 Wrong -> 40/60 Correct:",
    int((~correct_50 & correct_40).sum())
)
print(
    "50/50 Correct -> 40/60 Wrong:",
    int((correct_50 & ~correct_40).sum())
)

# ----------------------------------------------------------
# 8. Save reproducible results
# ----------------------------------------------------------

np.save(
    OUT_DIR / "v5_3seed_holdout_probs.npy",
    v5_new_probs
)

np.save(
    OUT_DIR / "go_holdout_probs.npy",
    go_probs
)

results_df.to_csv(
    OUT_DIR / "holdout_comparison.csv",
    index=False
)

pd.DataFrame(seed_results).to_csv(
    OUT_DIR / "seed_results.csv",
    index=False
)

print("\n========== COMPLETED ==========")
print("Saved:", OUT_DIR)
print("New holdout samples:", len(val_y))
print("Original V5 and GoEmotions models unchanged.")
print("Best Kaggle 0.6260 submission unchanged.")
print("No Kaggle submission performed.")
```

```python

# ==========================================================
# Experiment 41: Cross-Split Duplicate Audit
# Check duplicate texts between Train / Holdout / Old Val
# Prepare safe deduplication candidate (indices only)
# No training / No Kaggle submission
# ==========================================================

import re
import numpy as np
import pandas as pd
from pathlib import Path

SPLIT_DIR = Path("independent_holdout_2027")

OUT_DIR = Path(
    "training_data_quality_audit/split_overlap_audit"
)

if OUT_DIR.exists():
    raise FileExistsError(
        "Audit already exists. Previous results preserved."
    )

# ----------------------------------------------------------
# 1. Load existing split indices
# ----------------------------------------------------------

train_idx = np.load(
    SPLIT_DIR / "train_indices.npy"
)

holdout_idx = np.load(
    SPLIT_DIR / "validation_indices.npy"
)

old_val_idx = np.load(
    SPLIT_DIR / "old_validation_indices.npy"
)

assert len(train_idx) == 2150
assert len(holdout_idx) == 538
assert len(old_val_idx) == 672

# ----------------------------------------------------------
# 2. Conservative normalization
# ----------------------------------------------------------

def normalize_text(text):
    return re.sub(
        r"\s+", " ", str(text).lower()
    ).strip()

def prepare_split(indices):
    part = (
        train_df.iloc[indices]
        .copy()
        .reset_index(drop=True)
    )

    part["source_index"] = indices

    part["normalized_text"] = (
        part["text"]
        .fillna("")
        .map(normalize_text)
    )

    part["label"] = (
        part["label"]
        .astype(str)
        .str.lower()
        .str.strip()
    )

    return part

new_train = prepare_split(train_idx)
new_holdout = prepare_split(holdout_idx)
old_val = prepare_split(old_val_idx)

print("========== SPLIT SIZES ==========")
print("New Train:", len(new_train))
print("New Holdout:", len(new_holdout))
print("Old Validation:", len(old_val))

# ----------------------------------------------------------
# 3. Cross-split duplicate analysis
# ----------------------------------------------------------

def compare_splits(name_a, a, name_b, b):

    valid_a = a[a["normalized_text"] != ""]
    valid_b = b[b["normalized_text"] != ""]

    shared_texts = (
        set(valid_a["normalized_text"])
        & set(valid_b["normalized_text"])
    )

    overlapping_b = valid_b[
        valid_b["normalized_text"].isin(shared_texts)
    ].copy()

    labels_a = (
        valid_a.groupby("normalized_text")["label"]
        .agg(lambda x: set(x))
        .to_dict()
    )

    matching_label = sum(
        row.label in labels_a[row.normalized_text]
        for row in overlapping_b.itertuples()
    )

    conflicting_label = (
        len(overlapping_b) - matching_label
    )

    print(
        f"\n{name_a} vs {name_b}"
    )
    print("Shared text groups:", len(shared_texts))
    print(
        f"Affected {name_b} rows:",
        len(overlapping_b)
    )
    print("Labels matched:", matching_label)
    print("Labels conflicted:", conflicting_label)

    return {
        "Split A": name_a,
        "Split B": name_b,
        "Shared Text Groups": len(shared_texts),
        "Affected B Rows": len(overlapping_b),
        "Matching Labels": matching_label,
        "Conflicting Labels": conflicting_label
    }, overlapping_b

print("\n========== CROSS-SPLIT DUPLICATES ==========")

results = []

r1, overlap_holdout = compare_splits(
    "New Train", new_train,
    "New Holdout", new_holdout
)
results.append(r1)

r2, overlap_old = compare_splits(
    "New Train", new_train,
    "Old Validation", old_val
)
results.append(r2)

r3, overlap_between_val = compare_splits(
    "New Holdout", new_holdout,
    "Old Validation", old_val
)
results.append(r3)

# ----------------------------------------------------------
# 4. Deduplication candidates within New Training
# ----------------------------------------------------------

# Count same text + same label duplicates
same_label_duplicate_mask = (
    new_train.duplicated(
        subset=["normalized_text", "label"],
        keep="first"
    )
)

# Preserve different labels even if text matches
dedup_train = new_train[
    ~same_label_duplicate_mask
].copy()

# Detect text groups with conflicting labels
label_nunique = (
    new_train.groupby("normalized_text")["label"]
    .nunique()
)

conflict_groups = label_nunique[
    label_nunique > 1
]

print("\n========== TRAINING DEDUPLICATION PLAN ==========")

print("Original training rows:", len(new_train))

print(
    "Removable same-label duplicate rows:",
    int(same_label_duplicate_mask.sum())
)

print(
    "Training conflicting-label text groups:",
    len(conflict_groups)
)

print(
    "Training rows after safe deduplication:",
    len(dedup_train)
)

# Verify candidate still contains all six classes
assert (
    set(dedup_train["label"])
    == set(new_train["label"])
)

assert not dedup_train.duplicated(
    ["normalized_text", "label"]
).any()

# ----------------------------------------------------------
# 5. Print cross-split examples
# ----------------------------------------------------------

print("\n========== TRAIN/HOLDOUT OVERLAP EXAMPLES ==========")

if len(overlap_holdout):

    print(
        overlap_holdout[
            ["text", "label"]
        ]
        .head(12)
        .to_string(
            index=False,
            max_colwidth=100
        )
    )

else:
    print("No shared text found.")

# ----------------------------------------------------------
# 6. Save analysis and indices
# ----------------------------------------------------------

OUT_DIR.mkdir(parents=True, exist_ok=False)

pd.DataFrame(results).to_csv(
    OUT_DIR / "cross_split_summary.csv",
    index=False
)

overlap_holdout.to_csv(
    OUT_DIR / "train_holdout_overlap.csv",
    index=False,
    encoding="utf-8-sig"
)

overlap_old.to_csv(
    OUT_DIR / "train_oldval_overlap.csv",
    index=False,
    encoding="utf-8-sig"
)

# Only save source indices; do not modify train_df
np.save(
    OUT_DIR / "deduplicated_train_indices.npy",
    dedup_train["source_index"].to_numpy(dtype=int)
)

print("\n========== COMPLETED ==========")
print("Saved:", OUT_DIR)
print("Original training data unchanged.")
print("Existing models and submissions preserved.")
print("No training or Kaggle submission.")
```

```python

# ==========================================================
# Experiment 42: Remove Text-Overlap from Holdout Evaluation
# No training / No inference / No Kaggle submission
# ==========================================================

import re
import numpy as np
import pandas as pd

from pathlib import Path
from sklearn.metrics import (
    f1_score,
    accuracy_score,
    precision_recall_fscore_support
)

SPLIT_DIR = Path("independent_holdout_2027")

PROBS_DIR = (
    SPLIT_DIR / "distilbert_3seed_comparison"
)

OUT_DIR = Path(
    "training_data_quality_audit/"
    "split_overlap_audit/"
    "clean_holdout_evaluation"
)

if OUT_DIR.exists():
    raise FileExistsError(
        "Results already exist. Previous files preserved."
    )

EMOTIONS = [
    "anger", "disgust", "fear",
    "joy", "sadness", "surprise"
]

# ----------------------------------------------------------
# 1. Load saved indices and predictions
# ----------------------------------------------------------

train_idx = np.load(
    SPLIT_DIR / "train_indices.npy"
)

holdout_idx = np.load(
    SPLIT_DIR / "validation_indices.npy"
)

v5_probs = np.load(
    PROBS_DIR / "v5_3seed_holdout_probs.npy"
)

go_probs = np.load(
    PROBS_DIR / "go_holdout_probs.npy"
)

assert len(train_idx) == 2150
assert len(holdout_idx) == 538

assert v5_probs.shape == (538, 6)
assert go_probs.shape == (538, 6)

assert np.isfinite(v5_probs).all()
assert np.isfinite(go_probs).all()

# ----------------------------------------------------------
# 2. Rebuild text-overlap mask
# ----------------------------------------------------------

train_part = train_df.iloc[train_idx].copy()
val_part = train_df.iloc[holdout_idx].copy()

def normalize_text(text):
    return re.sub(
        r"\s+", " ",
        str(text).lower()
    ).strip()

train_texts = (
    train_part["text"]
    .fillna("")
    .map(normalize_text)
)

val_texts = (
    val_part["text"]
    .fillna("")
    .map(normalize_text)
)

train_text_set = set(
    train_texts[train_texts != ""]
)

overlap_mask = (
    val_texts.isin(train_text_set)
    & val_texts.ne("")
).to_numpy()

clean_mask = ~overlap_mask

print("========== HOLDOUT OVERLAP ==========")
print("Original Holdout:", len(val_part))
print("Overlapping rows:", int(overlap_mask.sum()))
print("Clean Holdout:", int(clean_mask.sum()))

assert overlap_mask.sum() == 18
assert clean_mask.sum() == 520

# ----------------------------------------------------------
# 3. Restore labels
# ----------------------------------------------------------

label_map = {
    name: i
    for i, name in enumerate(EMOTIONS)
}

if pd.api.types.is_numeric_dtype(val_part["label"]):

    y_true = val_part["label"].to_numpy(
        dtype=int
    )

else:

    y_true = (
        val_part["label"]
        .astype(str)
        .str.lower()
        .str.strip()
        .map(label_map)
        .to_numpy(dtype=int)
    )

assert len(y_true) == 538
assert np.isin(y_true, range(6)).all()

# ----------------------------------------------------------
# 4. Restore 50/50 and 40/60 predictions
# ----------------------------------------------------------

blend50 = (
    0.5 * v5_probs
    + 0.5 * go_probs
)

blend40 = (
    0.4 * v5_probs
    + 0.6 * go_probs
)

pred50 = blend50.argmax(axis=1)
pred40 = blend40.argmax(axis=1)

def evaluate(mask, pred):

    return (
        f1_score(
            y_true[mask],
            pred[mask],
            labels=range(6),
            average="macro",
            zero_division=0
        ),
        accuracy_score(
            y_true[mask],
            pred[mask]
        )
    )

full_mask = np.ones(
    len(y_true),
    dtype=bool
)

# Verify previous holdout scores
assert abs(
    evaluate(full_mask, pred50)[0] - 0.622470
) < 1e-5

assert abs(
    evaluate(full_mask, pred40)[0] - 0.626718
) < 1e-5

print("\nOriginal holdout results verified.")

# ----------------------------------------------------------
# 5. Evaluate original and clean holdout
# ----------------------------------------------------------

records = []

for name, mask in [
    ("Original Holdout", full_mask),
    ("Clean Holdout", clean_mask)
]:

    f1_50, acc_50 = evaluate(mask, pred50)
    f1_40, acc_40 = evaluate(mask, pred40)

    correct50 = pred50[mask] == y_true[mask]
    correct40 = pred40[mask] == y_true[mask]

    records.append({
        "Dataset": name,
        "Samples": int(mask.sum()),
        "50/50 F1": f1_50,
        "40/60 F1": f1_40,
        "Delta F1": f1_40 - f1_50,
        "50/50 Accuracy": acc_50,
        "40/60 Accuracy": acc_40,
        "Changed": int(
            (pred50[mask] != pred40[mask]).sum()
        ),
        "Fixed": int(
            (~correct50 & correct40).sum()
        ),
        "Broken": int(
            (correct50 & ~correct40).sum()
        )
    })

results_df = pd.DataFrame(records)

print("\n========== CLEAN HOLDOUT COMPARISON ==========")

print(
    results_df
    .round(6)
    .to_string(index=False)
)

# ----------------------------------------------------------
# 6. Examine excluded overlapping samples
# ----------------------------------------------------------

print("\n========== EXCLUDED 18 SAMPLES ==========")

for name, pred in [
    ("50/50", pred50),
    ("40/60", pred40)
]:

    correct = int(
        (
            pred[overlap_mask]
            == y_true[overlap_mask]
        ).sum()
    )

    print(
        f"{name}: {correct}/18 correct"
    )

# ----------------------------------------------------------
# 7. Per-class F1 on clean holdout
# ----------------------------------------------------------

def per_class_f1(pred):

    return precision_recall_fscore_support(
        y_true[clean_mask],
        pred[clean_mask],
        labels=range(6),
        zero_division=0
    )[2]

class_df = pd.DataFrame({
    "Emotion": EMOTIONS,
    "50/50 F1": per_class_f1(pred50),
    "40/60 F1": per_class_f1(pred40)
})

class_df["Difference"] = (
    class_df["40/60 F1"]
    - class_df["50/50 F1"]
)

print("\n========== CLEAN HOLDOUT PER-CLASS F1 ==========")

print(
    class_df
    .round(4)
    .to_string(index=False)
)

# ----------------------------------------------------------
# 8. Save results
# ----------------------------------------------------------

OUT_DIR.mkdir(
    parents=True,
    exist_ok=False
)

results_df.to_csv(
    OUT_DIR / "clean_holdout_comparison.csv",
    index=False
)

class_df.to_csv(
    OUT_DIR / "clean_holdout_per_class_f1.csv",
    index=False
)

np.save(
    OUT_DIR / "clean_holdout_mask.npy",
    clean_mask
)

print("\n========== COMPLETED ==========")
print("Saved:", OUT_DIR)
print("No training or inference.")
print("Original models and submissions preserved.")
```

```python

# ==========================================================
# Experiment 43: Twitter-RoBERTa-Large Emotion
# Zero-shot + fixed 20% ensemble
# Evaluate original validation and clean holdout
# No training / No Kaggle submission
# ==========================================================

import gc
import numpy as np
import pandas as pd
import torch

from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score, accuracy_score
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification
)

MODEL_NAME = "cardiffnlp/twitter-roberta-large-emotion-latest"

OUT_DIR = Path("twitter_roberta_large_emotion_validation")

if OUT_DIR.exists():
    raise FileExistsError("Results already exist. Preserved.")

EMOTIONS = [
    "anger", "disgust", "fear",
    "joy", "sadness", "surprise"
]

assert torch.cuda.is_available()

device = torch.device("cuda")

print("GPU:", torch.cuda.get_device_name(0))

# ----------------------------------------------------------
# 1. Restore original 672 validation samples
# ----------------------------------------------------------

_, old_val = train_test_split(
    train_df,
    test_size=0.2,
    random_state=42,
    stratify=train_df["label"]
)

old_val = old_val.reset_index(drop=True)

old_v5_probs = np.mean([
    np.load(
        "distilbert_ensemble_validation/"
        f"validation_probs_seed_{s}.npy"
    )
    for s in [42, 123, 2026]
], axis=0)

blend50 = np.load(
    "v5_goemotions_50_50_analysis/"
    "combined_validation_probs.npy"
)

old_go_probs = 2.0 * blend50 - old_v5_probs

old_baseline = (
    0.4 * old_v5_probs + 0.6 * old_go_probs
)

# ----------------------------------------------------------
# 2. Restore independent holdout predictions
# ----------------------------------------------------------

SPLIT_DIR = Path("independent_holdout_2027")
PROBS_DIR = SPLIT_DIR / "distilbert_3seed_comparison"

holdout_idx = np.load(
    SPLIT_DIR / "validation_indices.npy"
)

holdout = (
    train_df.iloc[holdout_idx]
    .reset_index(drop=True)
)

holdout_v5 = np.load(
    PROBS_DIR / "v5_3seed_holdout_probs.npy"
)

holdout_go = np.load(
    PROBS_DIR / "go_holdout_probs.npy"
)

holdout_baseline = (
    0.4 * holdout_v5 + 0.6 * holdout_go
)

clean_mask = np.load(
    "training_data_quality_audit/"
    "split_overlap_audit/"
    "clean_holdout_evaluation/"
    "clean_holdout_mask.npy"
)

assert clean_mask.sum() == 520

# ----------------------------------------------------------
# 3. Restore labels
# ----------------------------------------------------------

label_map = {
    e: i for i, e in enumerate(EMOTIONS)
}

def get_labels(df):
    if pd.api.types.is_numeric_dtype(df["label"]):
        return df["label"].to_numpy(dtype=int)

    return (
        df["label"].astype(str)
        .str.lower().str.strip()
        .map(label_map)
        .to_numpy(dtype=int)
    )

old_y = get_labels(old_val)
holdout_y = get_labels(holdout)

def score(y, pred):
    return (
        f1_score(
            y, pred,
            labels=range(6),
            average="macro",
            zero_division=0
        ),
        accuracy_score(y, pred)
    )

assert abs(
    score(old_y, old_baseline.argmax(axis=1))[0]
    - 0.667754
) < 1e-5

assert abs(
    score(
        holdout_y[clean_mask],
        holdout_baseline[clean_mask].argmax(axis=1)
    )[0] - 0.618655
) < 1e-5

print("Existing baselines verified.")

# ----------------------------------------------------------
# 4. Load Twitter-RoBERTa-Large
# ----------------------------------------------------------

print("\nLoading Twitter-RoBERTa-Large...")

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME
)

model_labels = {
    str(name).lower(): int(idx)
    for idx, name in model.config.id2label.items()
}

assert all(e in model_labels for e in EMOTIONS)

indices = [model_labels[e] for e in EMOTIONS]

print("Selected emotion indices:", indices)
print("Model output classes:", model.config.num_labels)

# FP16 inference to reduce GPU memory usage
model = model.to(
    device=device,
    dtype=torch.float16
)

model.eval()

# ----------------------------------------------------------
# 5. Inference function
# ----------------------------------------------------------

def predict_twitter(texts, batch_size=4):
    all_probs = []

    with torch.inference_mode():

        for start in range(0, len(texts), batch_size):

            batch = texts[start:start + batch_size]

            inputs = tokenizer(
                batch,
                padding=True,
                truncation=True,
                max_length=128,
                return_tensors="pt"
            ).to(device)

            logits = model(**inputs).logits.float()

            # Multi-label classifier -> sigmoid
            selected = torch.sigmoid(
                logits[:, indices]
            )

            # Normalize selected six emotions
            normalized = selected / selected.sum(
                dim=1, keepdim=True
            ).clamp_min(1e-12)

            all_probs.append(
                normalized.cpu().numpy()
            )

    return np.concatenate(all_probs, axis=0)

# ----------------------------------------------------------
# 6. Predict both evaluation sets
# ----------------------------------------------------------

print("\nPredicting old validation...")

old_twitter = predict_twitter(
    old_val["text"].fillna("").astype(str).tolist()
)

print("Predicting new holdout...")

holdout_twitter = predict_twitter(
    holdout["text"].fillna("").astype(str).tolist()
)

assert old_twitter.shape == (672, 6)
assert holdout_twitter.shape == (538, 6)

assert np.isfinite(old_twitter).all()
assert np.isfinite(holdout_twitter).all()

# ----------------------------------------------------------
# 7. Fixed 80/20 ensemble
# ----------------------------------------------------------

old_hybrid = (
    0.8 * old_baseline + 0.2 * old_twitter
)

holdout_hybrid = (
    0.8 * holdout_baseline + 0.2 * holdout_twitter
)

records = []

def add_result(dataset, y, base, twitter, hybrid):

    base_pred = base.argmax(axis=1)
    twitter_pred = twitter.argmax(axis=1)
    hybrid_pred = hybrid.argmax(axis=1)

    base_correct = base_pred == y
    hybrid_correct = hybrid_pred == y

    for name, pred in [
        ("40/60 Baseline", base_pred),
        ("Twitter-RoBERTa-Large", twitter_pred),
        ("80% Baseline + 20% Twitter", hybrid_pred)
    ]:

        f1, acc = score(y, pred)

        records.append({
            "Dataset": dataset,
            "Model": name,
            "Macro F1": f1,
            "Accuracy": acc,
            "Delta vs Baseline": (
                f1 - score(y, base_pred)[0]
            ),
            "Changed": int(
                (pred != base_pred).sum()
            ),
            "Fixed": int(
                (~base_correct & (pred == y)).sum()
            ),
            "Broken": int(
                (base_correct & (pred != y)).sum()
            )
        })

add_result(
    "Old Validation",
    old_y,
    old_baseline,
    old_twitter,
    old_hybrid
)

add_result(
    "Clean Holdout",
    holdout_y[clean_mask],
    holdout_baseline[clean_mask],
    holdout_twitter[clean_mask],
    holdout_hybrid[clean_mask]
)

results = pd.DataFrame(records)

print("\n========== MODEL COMPARISON ==========")
print(results.round(6).to_string(index=False))

# ----------------------------------------------------------
# 8. Save
# ----------------------------------------------------------

OUT_DIR.mkdir(parents=True, exist_ok=False)

results.to_csv(
    OUT_DIR / "model_comparison.csv",
    index=False
)

np.save(
    OUT_DIR / "old_validation_probs.npy",
    old_twitter
)

np.save(
    OUT_DIR / "holdout_probs.npy",
    holdout_twitter
)

del model, tokenizer
gc.collect()
torch.cuda.empty_cache()

print("\n========== COMPLETED ==========")
print("Saved:", OUT_DIR)
print("No training.")
print("No Kaggle submission.")
print("Current best 0.6260 preserved.")
```

```python

# ==========================================================
# Experiment 45: Supervised Emotion Adapter
# Frozen GoEmotions (28) + Frozen Twitter-Large (11)
# Train a lightweight 6-class classifier
# Compare on Old Validation and Clean Holdout
# No Transformer training / No Kaggle submission
# ==========================================================

import gc
import json
import joblib
import numpy as np
import pandas as pd
import torch

from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.metrics import f1_score, accuracy_score
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification
)

OUT_DIR = Path("emotion_adapter_39features")

if OUT_DIR.exists():
    raise FileExistsError(
        "Experiment already exists. Previous results preserved."
    )

EMOTIONS = [
    "anger", "disgust", "fear",
    "joy", "sadness", "surprise"
]

MODEL_SPECS = [
    ("SamLowe/roberta-base-go_emotions", 28, 16, False),
    ("cardiffnlp/twitter-roberta-large-emotion-latest",
     11, 4, True)
]

assert torch.cuda.is_available()

device = torch.device("cuda")

print("GPU:", torch.cuda.get_device_name(0))

# ----------------------------------------------------------
# 1. Restore both evaluation splits
# ----------------------------------------------------------

all_idx = np.arange(len(train_df))

dev_idx, old_val_idx = train_test_split(
    all_idx,
    test_size=0.2,
    random_state=42,
    stratify=train_df["label"]
)

split_dir = Path("independent_holdout_2027")

new_train_idx = np.load(
    split_dir / "train_indices.npy"
)

new_val_idx = np.load(
    split_dir / "validation_indices.npy"
)

clean_mask = np.load(
    "training_data_quality_audit/"
    "split_overlap_audit/"
    "clean_holdout_evaluation/"
    "clean_holdout_mask.npy"
)

assert len(dev_idx) == 2688
assert len(old_val_idx) == 672
assert len(new_train_idx) == 2150
assert len(new_val_idx) == 538
assert clean_mask.sum() == 520

label_map = {e: i for i, e in enumerate(EMOTIONS)}

if pd.api.types.is_numeric_dtype(train_df["label"]):
    y = train_df["label"].to_numpy(dtype=int)
else:
    y = (
        train_df["label"]
        .astype(str)
        .str.lower()
        .str.strip()
        .map(label_map)
        .to_numpy(dtype=int)
    )

assert np.isin(y, range(6)).all()

texts = (
    train_df["text"]
    .fillna("")
    .astype(str)
    .tolist()
)

# ----------------------------------------------------------
# 2. Extract all 39 frozen emotion features
# ----------------------------------------------------------

def extract_emotion_features(
    model_name, expected_labels,
    batch_size, use_fp16
):
    print("\nLoading:", model_name)

    tokenizer = AutoTokenizer.from_pretrained(
        model_name
    )

    model = AutoModelForSequenceClassification.from_pretrained(
        model_name
    )

    assert model.config.num_labels == expected_labels

    label_names = [
        str(model.config.id2label[i])
        for i in range(expected_labels)
    ]

    if use_fp16:
        model = model.to(
            device=device,
            dtype=torch.float16
        )
    else:
        model = model.to(device)

    model.eval()
    outputs = []

    with torch.inference_mode():

        for start in range(0, len(texts), batch_size):

            batch = texts[start:start + batch_size]

            inputs = tokenizer(
                batch,
                padding=True,
                truncation=True,
                max_length=128,
                return_tensors="pt"
            ).to(device)

            logits = model(**inputs).logits

            # Both are multilabel emotion classifiers
            scores = torch.sigmoid(
                logits.float()
            )

            outputs.append(
                scores.cpu().numpy()
            )

    features = np.concatenate(
        outputs, axis=0
    )

    assert features.shape == (
        len(texts), expected_labels
    )

    assert np.isfinite(features).all()

    print(
        "Extracted features:",
        features.shape
    )

    del model, tokenizer
    gc.collect()
    torch.cuda.empty_cache()

    return features, label_names

feature_blocks = []
all_feature_names = []

for name, n_labels, batch_size, use_fp16 in MODEL_SPECS:

    features, names = extract_emotion_features(
        name, n_labels, batch_size, use_fp16
    )

    feature_blocks.append(features)
    all_feature_names.extend(
        [f"{name.split('/')[0]}::{x}" for x in names]
    )

X = np.concatenate(
    feature_blocks,
    axis=1
).astype(np.float64)

assert X.shape == (3360, 39)

print("\n========== FEATURE MATRIX ==========")
print("Shape:", X.shape)

# ----------------------------------------------------------
# 3. Restore existing best ensemble probabilities
# ----------------------------------------------------------

old_v5 = np.mean([
    np.load(
        "distilbert_ensemble_validation/"
        f"validation_probs_seed_{s}.npy"
    )
    for s in [42, 123, 2026]
], axis=0)

old_blend50 = np.load(
    "v5_goemotions_50_50_analysis/"
    "combined_validation_probs.npy"
)

old_go = 2.0 * old_blend50 - old_v5

old_twitter = np.load(
    "twitter_roberta_large_emotion_validation/"
    "old_validation_probs.npy"
)

old_baseline = (
    0.32 * old_v5
    + 0.48 * old_go
    + 0.20 * old_twitter
)

holdout_dir = (
    split_dir / "distilbert_3seed_comparison"
)

holdout_v5 = np.load(
    holdout_dir / "v5_3seed_holdout_probs.npy"
)

holdout_go = np.load(
    holdout_dir / "go_holdout_probs.npy"
)

holdout_twitter = np.load(
    "twitter_roberta_large_emotion_validation/"
    "holdout_probs.npy"
)

holdout_baseline = (
    0.32 * holdout_v5
    + 0.48 * holdout_go
    + 0.20 * holdout_twitter
)

assert old_baseline.shape == (672, 6)
assert holdout_baseline.shape == (538, 6)

def macro_f1(true, pred):
    return f1_score(
        true, pred,
        labels=range(6),
        average="macro",
        zero_division=0
    )

assert abs(
    macro_f1(
        y[old_val_idx],
        old_baseline.argmax(axis=1)
    ) - 0.669120
) < 1e-5

assert abs(
    macro_f1(
        y[new_val_idx][clean_mask],
        holdout_baseline[clean_mask].argmax(axis=1)
    ) - 0.627932
) < 1e-5

print("Existing ensemble baselines verified.")

# ----------------------------------------------------------
# 4. Train adapters without evaluation label leakage
# ----------------------------------------------------------

def build_adapter():
    return make_pipeline(
        StandardScaler(),
        LogisticRegression(
            C=0.1,
            max_iter=3000,
            solver="lbfgs",
            random_state=123
        )
    )

# Adapter A: old development set only
adapter_old = build_adapter()

adapter_old.fit(
    X[dev_idx],
    y[dev_idx]
)

old_adapter_probs = adapter_old.predict_proba(
    X[old_val_idx]
)

# Adapter B: new training set only
adapter_new = build_adapter()

adapter_new.fit(
    X[new_train_idx],
    y[new_train_idx]
)

holdout_adapter_probs = adapter_new.predict_proba(
    X[new_val_idx]
)

assert np.array_equal(
    adapter_old.named_steps["logisticregression"].classes_,
    np.arange(6)
)

assert np.array_equal(
    adapter_new.named_steps["logisticregression"].classes_,
    np.arange(6)
)

# ----------------------------------------------------------
# 5. One fixed conservative blend
# ----------------------------------------------------------

# 80% current best + 20% supervised adapter
old_hybrid = (
    0.8 * old_baseline
    + 0.2 * old_adapter_probs
)

holdout_hybrid = (
    0.8 * holdout_baseline
    + 0.2 * holdout_adapter_probs
)

records = []

def compare(dataset_name, labels, baseline, adapter, hybrid):

    base_pred = baseline.argmax(axis=1)
    base_correct = base_pred == labels
    base_f1 = macro_f1(labels, base_pred)

    for name, probs in [
        ("Current Best", baseline),
        ("Supervised Adapter", adapter),
        ("80% Best + 20% Adapter", hybrid)
    ]:

        pred = probs.argmax(axis=1)
        correct = pred == labels

        records.append({
            "Dataset": dataset_name,
            "Model": name,
            "Macro F1": macro_f1(labels, pred),
            "Accuracy": accuracy_score(labels, pred),
            "Delta vs Best": macro_f1(labels, pred) - base_f1,
            "Changed": int((pred != base_pred).sum()),
            "Fixed": int((~base_correct & correct).sum()),
            "Broken": int((base_correct & ~correct).sum())
        })

compare(
    "Old Validation",
    y[old_val_idx],
    old_baseline,
    old_adapter_probs,
    old_hybrid
)

compare(
    "Clean Holdout",
    y[new_val_idx][clean_mask],
    holdout_baseline[clean_mask],
    holdout_adapter_probs[clean_mask],
    holdout_hybrid[clean_mask]
)

results = pd.DataFrame(records)

print("\n========== ADAPTER COMPARISON ==========")
print(
    results.round(6).to_string(index=False)
)

# ----------------------------------------------------------
# 6. Save all diagnostics
# ----------------------------------------------------------

OUT_DIR.mkdir(parents=True, exist_ok=False)

results.to_csv(
    OUT_DIR / "adapter_comparison.csv",
    index=False
)

np.savez_compressed(
    OUT_DIR / "emotion_features.npz",
    features=X.astype(np.float32),
    labels=y
)

joblib.dump(
    adapter_old,
    OUT_DIR / "adapter_old_validation.joblib"
)

joblib.dump(
    adapter_new,
    OUT_DIR / "adapter_clean_holdout.joblib"
)

with open(
    OUT_DIR / "feature_names.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        all_feature_names,
        f,
        ensure_ascii=False,
        indent=2
    )

print("\n========== COMPLETED ==========")
print("Saved:", OUT_DIR)
print("Current 0.6289 model unchanged.")
print("No Transformer training.")
print("No Kaggle submission.")
```

```python

# ==========================================================
# Experiment 46 - Step 4
# Twitter-RoBERTa-Large LoRA: Train 1 Epoch
#
# Train: 2150 samples (new training split)
# Evaluate: Old Validation 672 + Clean Holdout 520
# Preserve current Kaggle best 0.6289
# ==========================================================

import gc
import json
import math
import random
import numpy as np
import pandas as pd
import torch
import torch.nn.functional as F

from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score, accuracy_score
from torch.utils.data import Dataset, DataLoader
from tqdm.auto import tqdm

# ----------------------------------------------------------
# 0. Configuration
# ----------------------------------------------------------

SEED = 123
EPOCHS = 1
BATCH_SIZE = 4
ACCUM_STEPS = 4
MAX_LENGTH = 128
LORA_LR = 1e-4
HEAD_LR = 2e-5

OUT_DIR = Path(
    "twitter_roberta_large_lora_2150_epoch1"
)

assert not OUT_DIR.exists(), (
    "Output already exists. Previous experiment preserved."
)

assert "lora_model" in globals(), (
    "Run Experiment 46 Step 3 first."
)

assert "lora_tokenizer" in globals()

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

device = torch.device("cuda")

EMOTIONS = [
    "anger", "disgust", "fear",
    "joy", "sadness", "surprise"
]

label_map = {
    emotion: i for i, emotion in enumerate(EMOTIONS)
}

# ----------------------------------------------------------
# 1. Restore data splits
# ----------------------------------------------------------

all_indices = np.arange(len(train_df))

_, old_val_idx = train_test_split(
    all_indices,
    test_size=0.2,
    random_state=42,
    stratify=train_df["label"]
)

split_dir = Path("independent_holdout_2027")

train_idx = np.load(
    split_dir / "train_indices.npy"
)

holdout_idx = np.load(
    split_dir / "validation_indices.npy"
)

clean_mask = np.load(
    "training_data_quality_audit/"
    "split_overlap_audit/"
    "clean_holdout_evaluation/"
    "clean_holdout_mask.npy"
).astype(bool)

assert len(train_idx) == 2150
assert len(holdout_idx) == 538
assert len(old_val_idx) == 672
assert clean_mask.sum() == 520

assert not np.intersect1d(
    train_idx, holdout_idx
).size

assert not np.intersect1d(
    train_idx, old_val_idx
).size

if pd.api.types.is_numeric_dtype(train_df["label"]):
    y = train_df["label"].to_numpy(dtype=int)
else:
    y = (
        train_df["label"]
        .astype(str)
        .str.lower()
        .str.strip()
        .map(label_map)
        .to_numpy(dtype=int)
    )

assert np.isin(y, range(6)).all()

texts = (
    train_df["text"]
    .fillna("")
    .astype(str)
    .tolist()
)

print("========== DATA ==========")
print("Train:", len(train_idx))
print("Old Validation:", len(old_val_idx))
print("Clean Holdout:", clean_mask.sum())

# ----------------------------------------------------------
# 2. Tokenize once
# ----------------------------------------------------------

tokens = lora_tokenizer(
    texts,
    truncation=True,
    max_length=MAX_LENGTH,
    padding=False
)

class EmotionDataset(Dataset):

    def __init__(self, indices):
        self.indices = np.asarray(indices)

    def __len__(self):
        return len(self.indices)

    def __getitem__(self, i):

        idx = int(self.indices[i])

        return {
            "input_ids": tokens["input_ids"][idx],
            "attention_mask": tokens["attention_mask"][idx],
            "label": int(y[idx])
        }

def collate_fn(batch):

    inputs = lora_tokenizer.pad(
        [
            {
                "input_ids": x["input_ids"],
                "attention_mask": x["attention_mask"]
            }
            for x in batch
        ],
        padding=True,
        return_tensors="pt"
    )

    labels = torch.tensor(
        [x["label"] for x in batch],
        dtype=torch.long
    )

    return inputs, labels

def make_loader(indices, batch_size, shuffle=False):

    return DataLoader(
        EmotionDataset(indices),
        batch_size=batch_size,
        shuffle=shuffle,
        collate_fn=collate_fn,
        num_workers=0,
        generator=torch.Generator().manual_seed(SEED)
    )

train_loader = make_loader(
    train_idx,
    batch_size=BATCH_SIZE,
    shuffle=True
)

old_val_loader = make_loader(
    old_val_idx,
    batch_size=8
)

holdout_loader = make_loader(
    holdout_idx,
    batch_size=8
)

# ----------------------------------------------------------
# 3. Prepare trainable parameters
# ----------------------------------------------------------

lora_model.to(device)

lora_model.gradient_checkpointing_enable()
lora_model.enable_input_require_grads()
lora_model.config.use_cache = False

lora_params = []
head_params = []

for name, p in lora_model.named_parameters():

    if not p.requires_grad:
        continue

    if "classifier" in name:

        # Keep the trainable head in FP32.
        # Autocast still uses BF16 for suitable operations.
        p.data = p.data.float()
        head_params.append(p)

    else:
        lora_params.append(p)

assert len(lora_params) > 0
assert len(head_params) > 0

optimizer = torch.optim.AdamW(
    [
        {
            "params": lora_params,
            "lr": LORA_LR
        },
        {
            "params": head_params,
            "lr": HEAD_LR
        }
    ],
    weight_decay=0.01
)

lora_model.zero_grad(set_to_none=True)

selected_indices = [
    int(lora_model.config.label2id[e])
    for e in EMOTIONS
] if all(
    e in lora_model.config.label2id
    for e in EMOTIONS
) else [
    {
        str(name).lower(): int(i)
        for i, name in
        lora_model.config.id2label.items()
    }[e]
    for e in EMOTIONS
]

assert len(selected_indices) == 6

print("\n========== TRAINING CONFIG ==========")
print("Epochs:", EPOCHS)
print("Batch:", BATCH_SIZE)
print("Gradient accumulation:", ACCUM_STEPS)
print("Effective batch:", BATCH_SIZE * ACCUM_STEPS)
print("LoRA LR:", LORA_LR)
print("Head LR:", HEAD_LR)
print("Selected indices:", selected_indices)

# ----------------------------------------------------------
# 4. Train exactly one epoch
# ----------------------------------------------------------

print("\n========== LORA TRAINING ==========")

lora_model.train()

torch.cuda.reset_peak_memory_stats()

total_loss = 0.0
seen = 0

optimizer.zero_grad(set_to_none=True)

progress = tqdm(
    enumerate(train_loader),
    total=len(train_loader),
    desc="Epoch 1/1"
)

for step, (inputs, labels) in progress:

    inputs = {
        k: v.to(device)
        for k, v in inputs.items()
    }

    labels = labels.to(device)

    group_start = (
        step // ACCUM_STEPS
    ) * ACCUM_STEPS

    group_size = min(
        ACCUM_STEPS,
        len(train_loader) - group_start
    )

    with torch.autocast(
        device_type="cuda",
        dtype=torch.bfloat16
    ):

        logits = lora_model(**inputs).logits

        six_logits = logits[
            :, selected_indices
        ].float()

        raw_loss = F.cross_entropy(
            six_logits,
            labels
        )

        loss = raw_loss / group_size

    assert torch.isfinite(raw_loss).item(), (
        "Non-finite training loss."
    )

    loss.backward()

    total_loss += (
        raw_loss.item() * len(labels)
    )

    seen += len(labels)

    should_update = (
        (step + 1) % ACCUM_STEPS == 0
        or (step + 1) == len(train_loader)
    )

    if should_update:

        torch.nn.utils.clip_grad_norm_(
            [
                p for p in
                lora_model.parameters()
                if p.requires_grad
            ],
            max_norm=1.0,
            error_if_nonfinite=True
        )

        optimizer.step()
        optimizer.zero_grad(set_to_none=True)

    if (step + 1) % 25 == 0:
        progress.set_postfix(
            loss=f"{total_loss / seen:.4f}"
        )

train_loss = total_loss / seen

print("\nTraining completed.")
print("Train samples:", seen)
print("Average train loss:", round(train_loss, 6))
print(
    "Peak allocated VRAM:",
    f"{torch.cuda.max_memory_allocated() / 1024**3:.3f} GB"
)

# ----------------------------------------------------------
# 5. Predict using the trained LoRA model
# ----------------------------------------------------------

def predict_lora(loader):

    lora_model.eval()

    predictions = []

    with torch.inference_mode():

        for inputs, labels in tqdm(
            loader,
            desc="Evaluation",
            leave=False
        ):

            inputs = {
                k: v.to(device)
                for k, v in inputs.items()
            }

            with torch.autocast(
                device_type="cuda",
                dtype=torch.bfloat16
            ):

                logits = (
                    lora_model(**inputs)
                    .logits.float()
                )

            # Match the original Twitter model's
            # sigmoid + 6-class normalization
            scores = torch.sigmoid(
                logits[:, selected_indices]
            )

            probs = (
                scores
                / scores.sum(
                    dim=1,
                    keepdim=True
                ).clamp_min(1e-12)
            )

            predictions.append(
                probs.cpu().numpy()
            )

    return np.concatenate(
        predictions,
        axis=0
    )

print("\n========== VALIDATION INFERENCE ==========")

old_lora = predict_lora(old_val_loader)
holdout_lora = predict_lora(holdout_loader)

assert old_lora.shape == (672, 6)
assert holdout_lora.shape == (538, 6)

assert np.isfinite(old_lora).all()
assert np.isfinite(holdout_lora).all()

# ----------------------------------------------------------
# 6. Restore the current best three-model ensemble
# ----------------------------------------------------------

old_v5 = np.mean([
    np.load(
        "distilbert_ensemble_validation/"
        f"validation_probs_seed_{s}.npy"
    )
    for s in [42, 123, 2026]
], axis=0)

old_blend50 = np.load(
    "v5_goemotions_50_50_analysis/"
    "combined_validation_probs.npy"
)

old_go = 2 * old_blend50 - old_v5

old_twitter = np.load(
    "twitter_roberta_large_emotion_validation/"
    "old_validation_probs.npy"
)

holdout_dir = (
    split_dir / "distilbert_3seed_comparison"
)

holdout_v5 = np.load(
    holdout_dir / "v5_3seed_holdout_probs.npy"
)

holdout_go = np.load(
    holdout_dir / "go_holdout_probs.npy"
)

holdout_twitter = np.load(
    "twitter_roberta_large_emotion_validation/"
    "holdout_probs.npy"
)

# Original best: V5 32%, GO 48%, Twitter 20%
old_best = (
    0.32 * old_v5
    + 0.48 * old_go
    + 0.20 * old_twitter
)

holdout_best = (
    0.32 * holdout_v5
    + 0.48 * holdout_go
    + 0.20 * holdout_twitter
)

# Candidate A: Replace Twitter entirely with LoRA
old_replace = (
    0.32 * old_v5
    + 0.48 * old_go
    + 0.20 * old_lora
)

holdout_replace = (
    0.32 * holdout_v5
    + 0.48 * holdout_go
    + 0.20 * holdout_lora
)

# Candidate B: Preserve half of original Twitter
old_mixed = (
    0.32 * old_v5
    + 0.48 * old_go
    + 0.10 * old_twitter
    + 0.10 * old_lora
)

holdout_mixed = (
    0.32 * holdout_v5
    + 0.48 * holdout_go
    + 0.10 * holdout_twitter
    + 0.10 * holdout_lora
)

# ----------------------------------------------------------
# 7. Compare predictions
# ----------------------------------------------------------

def macro_f1(y_true, pred):

    return f1_score(
        y_true,
        pred,
        labels=range(6),
        average="macro",
        zero_division=0
    )

old_y = y[old_val_idx]
holdout_y = y[holdout_idx][clean_mask]

assert abs(
    macro_f1(
        old_y,
        old_best.argmax(axis=1)
    ) - 0.669120
) < 1e-5

assert abs(
    macro_f1(
        holdout_y,
        holdout_best[clean_mask].argmax(axis=1)
    ) - 0.627932
) < 1e-5

records = []

def add_results(
    dataset, labels, baseline,
    lora_probs, replaced, mixed
):

    base_pred = baseline.argmax(axis=1)
    base_correct = base_pred == labels
    base_f1 = macro_f1(labels, base_pred)

    candidates = [
        ("Current Best 0.6289", baseline),
        ("LoRA Alone", lora_probs),
        ("Replace Twitter with LoRA", replaced),
        ("10% Twitter + 10% LoRA", mixed)
    ]

    for name, probs in candidates:

        pred = probs.argmax(axis=1)
        correct = pred == labels

        records.append({
            "Dataset": dataset,
            "Model": name,
            "Macro F1": macro_f1(labels, pred),
            "Accuracy": accuracy_score(labels, pred),
            "Delta vs Best": (
                macro_f1(labels, pred) - base_f1
            ),
            "Changed": int(
                (pred != base_pred).sum()
            ),
            "Fixed": int(
                (~base_correct & correct).sum()
            ),
            "Broken": int(
                (base_correct & ~correct).sum()
            )
        })

add_results(
    "Old Validation",
    old_y,
    old_best,
    old_lora,
    old_replace,
    old_mixed
)

add_results(
    "Clean Holdout",
    holdout_y,
    holdout_best[clean_mask],
    holdout_lora[clean_mask],
    holdout_replace[clean_mask],
    holdout_mixed[clean_mask]
)

results = pd.DataFrame(records)

print("\n========== LORA MODEL COMPARISON ==========")
print(results.round(6).to_string(index=False))

# ----------------------------------------------------------
# 8. Save adapter and experiment outputs
# ----------------------------------------------------------

OUT_DIR.mkdir(parents=True, exist_ok=False)

adapter_dir = OUT_DIR / "adapter"

lora_model.save_pretrained(
    adapter_dir,
    safe_serialization=True
)

lora_tokenizer.save_pretrained(adapter_dir)

np.savez_compressed(
    OUT_DIR / "validation_predictions.npz",
    old_lora_probs=old_lora,
    holdout_lora_probs=holdout_lora,
    old_best_probs=old_best,
    holdout_best_probs=holdout_best,
    old_validation_indices=old_val_idx,
    holdout_indices=holdout_idx,
    clean_mask=clean_mask
)

results.to_csv(
    OUT_DIR / "lora_comparison.csv",
    index=False
)

config = {
    "base_model":
        "cardiffnlp/twitter-roberta-large-emotion-latest",
    "train_samples": int(len(train_idx)),
    "epochs": EPOCHS,
    "batch_size": BATCH_SIZE,
    "gradient_accumulation": ACCUM_STEPS,
    "max_length": MAX_LENGTH,
    "lora_lr": LORA_LR,
    "head_lr": HEAD_LR,
    "train_loss": float(train_loss),
    "seed": SEED,
    "selected_indices": selected_indices
}

with open(
    OUT_DIR / "training_config.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(config, f, indent=2)

print("\n========== COMPLETED ==========")
print("Adapter saved:", adapter_dir)
print("Results saved:", OUT_DIR)
print("No Kaggle submission generated.")
print("Current best Public F1 0.6289 preserved.")
```
